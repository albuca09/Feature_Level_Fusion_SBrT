```python
from pathlib import Path
import zipfile
import pandas as pd
from collections import Counter, defaultdict

# ============================================================
# CONFIGURAÇÃO
# ============================================================

ZIP_PATH = Path(r"C:\Users\Luis\Downloads\Drone-detection-dataset-master.zip")

OUT_DIR = ZIP_PATH.parent / "exploracao_drone_dataset"
OUT_DIR.mkdir(parents=True, exist_ok=True)

CSV_FILES = OUT_DIR / "arquivos_no_zip.csv"
CSV_FOLDERS = OUT_DIR / "pastas_no_zip.csv"
TXT_TREE = OUT_DIR / "estrutura_pastas.txt"

# ============================================================
# VERIFICAÇÃO
# ============================================================

if not ZIP_PATH.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {ZIP_PATH}")

if ZIP_PATH.suffix.lower() != ".zip":
    raise ValueError(f"O caminho informado não é um ZIP: {ZIP_PATH}")

print(f"[OK] ZIP encontrado: {ZIP_PATH}")
print(f"[INFO] Resultados serão salvos em: {OUT_DIR}")

# ============================================================
# LER CONTEÚDO DO ZIP SEM EXTRAIR
# ============================================================

files_data = []
folders_set = set()

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    infos = z.infolist()

    for info in infos:
        path = Path(info.filename)

        # Ignora entradas vazias
        if str(path).strip() == "":
            continue

        # Pasta
        if info.is_dir():
            folders_set.add(str(path).replace("\\", "/"))
            continue

        # Arquivo
        parts = path.parts
        folder = str(Path(*parts[:-1])).replace("\\", "/") if len(parts) > 1 else ""
        filename = path.name
        suffix = path.suffix.lower()

        folders_set.add(folder)

        files_data.append({
            "zip_internal_path": info.filename,
            "folder": folder,
            "filename": filename,
            "extension": suffix,
            "size_bytes": info.file_size,
            "size_mb": round(info.file_size / (1024 * 1024), 4),
            "compressed_size_bytes": info.compress_size,
        })

# ============================================================
# DATAFRAMES
# ============================================================

df_files = pd.DataFrame(files_data)

df_folders = pd.DataFrame({
    "folder": sorted(folders_set)
})

# Profundidade das pastas
df_folders["depth"] = df_folders["folder"].apply(
    lambda x: 0 if x == "" else len(Path(x).parts)
)

# ============================================================
# RESUMOS
# ============================================================

print("\n" + "=" * 80)
print("RESUMO DO ZIP")
print("=" * 80)

print(f"Total de arquivos : {len(df_files)}")
print(f"Total de pastas   : {len(df_folders)}")

if not df_files.empty:
    print(f"Tamanho total     : {df_files['size_mb'].sum():.2f} MB")

    print("\nExtensões encontradas:")
    ext_counts = df_files["extension"].value_counts(dropna=False)
    print(ext_counts)

    print("\nTop 20 maiores arquivos:")
    print(
        df_files.sort_values("size_bytes", ascending=False)
        [["zip_internal_path", "size_mb"]]
        .head(20)
        .to_string(index=False)
    )

# ============================================================
# GERAR ÁRVORE DE PASTAS
# ============================================================

def build_tree(paths):
    tree = defaultdict(dict)

    for p in paths:
        if not p:
            continue

        parts = Path(p).parts
        current = tree

        for part in parts:
            current = current.setdefault(part, {})

    return tree


def write_tree(tree, file, indent=0):
    for key in sorted(tree.keys()):
        file.write("    " * indent + f"📁 {key}\n")
        write_tree(tree[key], file, indent + 1)


folder_tree = build_tree(df_folders["folder"].tolist())

with open(TXT_TREE, "w", encoding="utf-8") as f:
    f.write(f"Estrutura de pastas do ZIP:\n{ZIP_PATH}\n\n")
    write_tree(folder_tree, f)

# ============================================================
# SALVAR RESULTADOS
# ============================================================

df_files.to_csv(CSV_FILES, index=False, encoding="utf-8-sig")
df_folders.to_csv(CSV_FOLDERS, index=False, encoding="utf-8-sig")

print("\n" + "=" * 80)
print("ARQUIVOS SALVOS")
print("=" * 80)

print(f"[OK] Lista de arquivos : {CSV_FILES}")
print(f"[OK] Lista de pastas   : {CSV_FOLDERS}")
print(f"[OK] Árvore de pastas  : {TXT_TREE}")

# ============================================================
# MOSTRAR PRIMEIRAS LINHAS
# ============================================================

print("\nPrimeiros arquivos encontrados:")
print(df_files.head(30).to_string(index=False))
```

```python
from pathlib import Path
import pandas as pd
import re

# ============================================================
# CAMINHOS
# ============================================================

CSV_FILES = Path(r"C:\Users\Luis\Downloads\exploracao_drone_dataset\arquivos_no_zip.csv")

OUT_DIR = CSV_FILES.parent
OUT_SUMMARY = OUT_DIR / "resumo_por_pasta_classe.csv"
OUT_CLASSES = OUT_DIR / "resumo_classes.csv"

df = pd.read_csv(CSV_FILES)

# ============================================================
# FUNÇÕES AUXILIARES
# ============================================================

def infer_modality(folder, filename, extension):
    folder = str(folder).lower()
    filename = str(filename).lower()

    if extension == ".wav":
        return "audio"
    if extension == ".mp4" and "video_ir" in folder:
        return "video_ir"
    if extension == ".mp4" and "video_v" in folder:
        return "video_visible"
    if extension == ".mat":
        return "matlab_signal"
    if extension == ".xlsx":
        return "spreadsheet"
    if extension == ".md":
        return "readme"
    return "other"


def infer_class(filename):
    """
    Extrai a classe a partir do padrão do nome:
    DRONE_001.wav
    BACKGROUND_001.wav
    V_BIRD_005.mp4
    IR_AIRPLANE_072.mp4
    etc.
    """
    name = Path(str(filename)).stem.upper()

    # Remove prefixos de modalidade
    name = re.sub(r"^(V_|IR_)", "", name)

    # Remove numeração final
    name = re.sub(r"_\d+$", "", name)

    return name


df["modality"] = df.apply(
    lambda r: infer_modality(r["folder"], r["filename"], r["extension"]),
    axis=1
)

df["class"] = df["filename"].apply(infer_class)

# ============================================================
# RESUMO POR PASTA / MODALIDADE / EXTENSÃO / CLASSE
# ============================================================

summary = (
    df.groupby(["folder", "modality", "extension", "class"])
    .agg(
        n_files=("filename", "count"),
        total_mb=("size_mb", "sum"),
        mean_mb=("size_mb", "mean"),
    )
    .reset_index()
    .sort_values(["folder", "modality", "class"])
)

summary["total_mb"] = summary["total_mb"].round(3)
summary["mean_mb"] = summary["mean_mb"].round(3)

summary.to_csv(OUT_SUMMARY, index=False, encoding="utf-8-sig")

# ============================================================
# RESUMO GLOBAL POR CLASSE
# ============================================================

class_summary = (
    df.groupby(["class", "modality", "extension"])
    .agg(
        n_files=("filename", "count"),
        total_mb=("size_mb", "sum"),
    )
    .reset_index()
    .sort_values(["class", "modality"])
)

class_summary["total_mb"] = class_summary["total_mb"].round(3)

class_summary.to_csv(OUT_CLASSES, index=False, encoding="utf-8-sig")

# ============================================================
# EXIBIR RESULTADOS
# ============================================================

print("\n" + "=" * 80)
print("RESUMO POR MODALIDADE")
print("=" * 80)
print(
    df.groupby(["modality", "extension"])
    .agg(n_files=("filename", "count"), total_mb=("size_mb", "sum"))
    .reset_index()
    .sort_values("n_files", ascending=False)
    .to_string(index=False)
)

print("\n" + "=" * 80)
print("RESUMO POR CLASSE")
print("=" * 80)
print(
    df.groupby("class")
    .agg(n_files=("filename", "count"), total_mb=("size_mb", "sum"))
    .reset_index()
    .sort_values("n_files", ascending=False)
    .to_string(index=False)
)

print("\n" + "=" * 80)
print("ARQUIVOS SALVOS")
print("=" * 80)
print(f"[OK] Resumo por pasta/classe: {OUT_SUMMARY}")
print(f"[OK] Resumo por classe      : {OUT_CLASSES}")
```

```python
from pathlib import Path
import pandas as pd
import re

# ============================================================
# CAMINHOS
# ============================================================

CSV_FILES = Path(r"C:\Users\Luis\Downloads\exploracao_drone_dataset\arquivos_no_zip.csv")

OUT_DIR = CSV_FILES.parent
OUT_MANIFEST = OUT_DIR / "dataset_manifest_limpo.csv"
OUT_CLASS_SUMMARY = OUT_DIR / "resumo_classes_limpo.csv"
OUT_PAIR_SUMMARY = OUT_DIR / "resumo_pares_video_label.csv"
OUT_MISSING = OUT_DIR / "arquivos_sem_par.csv"

df = pd.read_csv(CSV_FILES)

# ============================================================
# FUNÇÕES
# ============================================================

def clean_stem(filename):
    return Path(str(filename)).stem.upper()


def detect_modality(row):
    folder = str(row["folder"]).lower()
    filename = str(row["filename"]).lower()
    ext = str(row["extension"]).lower()

    if ext == ".wav":
        return "audio"

    if ext == ".mp4":
        if "video_ir" in folder:
            return "video_ir"
        if "video_v" in folder:
            return "video_visible"
        return "video_other"

    if ext == ".mat":
        return "label_mat"

    if ext == ".xlsx":
        return "spreadsheet"

    if ext == ".md":
        return "readme"

    if ext == ".m":
        return "matlab_script"

    return "other"


def normalize_sample_name(filename):
    """
    Exemplos:
    V_DRONE_001.mp4         -> DRONE_001
    IR_DRONE_001.mp4        -> DRONE_001
    DRONE_001_LABELS.mat    -> DRONE_001
    DRONE_001.wav           -> DRONE_001
    """
    stem = clean_stem(filename)

    # Remove prefixo de modalidade
    stem = re.sub(r"^(V_|IR_)", "", stem)

    # Remove sufixo de label
    stem = re.sub(r"_LABELS$", "", stem)

    return stem


def extract_object_class(sample_name):
    """
    DRONE_001      -> DRONE
    BIRD_020       -> BIRD
    AIRPLANE_072   -> AIRPLANE
    HELICOPTER_061 -> HELICOPTER
    BACKGROUND_001 -> BACKGROUND
    """
    sample_name = str(sample_name).upper()

    match = re.match(r"^([A-Z]+)_\d+$", sample_name)

    if match:
        return match.group(1)

    return sample_name


def extract_sample_number(sample_name):
    match = re.search(r"_(\d+)$", str(sample_name))
    if match:
        return int(match.group(1))
    return None


# ============================================================
# PROCESSAMENTO
# ============================================================

df["modality_clean"] = df.apply(detect_modality, axis=1)
df["sample_name"] = df["filename"].apply(normalize_sample_name)
df["object_class"] = df["sample_name"].apply(extract_object_class)
df["sample_number"] = df["sample_name"].apply(extract_sample_number)

# Remove arquivos auxiliares do resumo principal
valid_modalities = [
    "audio",
    "video_ir",
    "video_visible",
    "label_mat",
    "video_other",
]

df_clean = df[df["modality_clean"].isin(valid_modalities)].copy()

# ============================================================
# RESUMO LIMPO POR CLASSE E MODALIDADE
# ============================================================

class_summary = (
    df_clean
    .groupby(["object_class", "modality_clean", "extension"])
    .agg(
        n_files=("filename", "count"),
        total_mb=("size_mb", "sum"),
        mean_mb=("size_mb", "mean")
    )
    .reset_index()
    .sort_values(["object_class", "modality_clean"])
)

class_summary["total_mb"] = class_summary["total_mb"].round(3)
class_summary["mean_mb"] = class_summary["mean_mb"].round(3)

# ============================================================
# MANIFESTO POR AMOSTRA
# ============================================================

manifest_rows = []

for sample_name, g in df_clean.groupby("sample_name"):
    row = {
        "sample_name": sample_name,
        "object_class": extract_object_class(sample_name),
        "sample_number": extract_sample_number(sample_name),
        "has_audio": False,
        "has_video_ir": False,
        "has_video_visible": False,
        "has_label_mat": False,
        "audio_path": "",
        "video_ir_path": "",
        "video_visible_path": "",
        "label_mat_paths": "",
        "n_label_mat": 0,
    }

    labels = []

    for _, r in g.iterrows():
        mod = r["modality_clean"]
        path = r["zip_internal_path"]

        if mod == "audio":
            row["has_audio"] = True
            row["audio_path"] = path

        elif mod == "video_ir":
            row["has_video_ir"] = True
            row["video_ir_path"] = path

        elif mod == "video_visible":
            row["has_video_visible"] = True
            row["video_visible_path"] = path

        elif mod == "label_mat":
            row["has_label_mat"] = True
            labels.append(path)

    row["n_label_mat"] = len(labels)
    row["label_mat_paths"] = " | ".join(labels)

    manifest_rows.append(row)

manifest = pd.DataFrame(manifest_rows)

manifest = manifest.sort_values(
    ["object_class", "sample_number"],
    na_position="last"
)

# ============================================================
# RESUMO DE PARES VÍDEO-LABEL
# ============================================================

pair_summary = manifest.copy()

pair_summary["has_any_video"] = (
    pair_summary["has_video_ir"] | pair_summary["has_video_visible"]
)

pair_summary["video_label_status"] = "no_video"

pair_summary.loc[
    pair_summary["has_any_video"] & pair_summary["has_label_mat"],
    "video_label_status"
] = "video_with_label"

pair_summary.loc[
    pair_summary["has_any_video"] & ~pair_summary["has_label_mat"],
    "video_label_status"
] = "video_without_label"

pair_summary.loc[
    ~pair_summary["has_any_video"] & pair_summary["has_label_mat"],
    "video_label_status"
] = "label_without_video"

pair_status_summary = (
    pair_summary
    .groupby(["object_class", "video_label_status"])
    .agg(n_samples=("sample_name", "count"))
    .reset_index()
    .sort_values(["object_class", "video_label_status"])
)

# Arquivos/amostras problemáticas
missing = pair_summary[
    pair_summary["video_label_status"].isin(
        ["video_without_label", "label_without_video"]
    )
].copy()

# ============================================================
# SALVAR
# ============================================================

df_clean.to_csv(OUT_MANIFEST, index=False, encoding="utf-8-sig")
class_summary.to_csv(OUT_CLASS_SUMMARY, index=False, encoding="utf-8-sig")
pair_status_summary.to_csv(OUT_PAIR_SUMMARY, index=False, encoding="utf-8-sig")
missing.to_csv(OUT_MISSING, index=False, encoding="utf-8-sig")

# ============================================================
# EXIBIR RESULTADOS
# ============================================================

print("\n" + "=" * 80)
print("RESUMO LIMPO POR CLASSE / MODALIDADE")
print("=" * 80)
print(class_summary.to_string(index=False))

print("\n" + "=" * 80)
print("RESUMO DE PARES VÍDEO-LABEL")
print("=" * 80)
print(pair_status_summary.to_string(index=False))

print("\n" + "=" * 80)
print("AMOSTRAS COM PROBLEMA DE PAREAMENTO")
print("=" * 80)

if missing.empty:
    print("[OK] Nenhuma amostra com vídeo sem label ou label sem vídeo.")
else:
    print(missing.to_string(index=False))

print("\n" + "=" * 80)
print("ARQUIVOS SALVOS")
print("=" * 80)
print(f"[OK] Manifesto limpo          : {OUT_MANIFEST}")
print(f"[OK] Resumo classes limpo    : {OUT_CLASS_SUMMARY}")
print(f"[OK] Resumo pares vídeo-label: {OUT_PAIR_SUMMARY}")
print(f"[OK] Arquivos sem par        : {OUT_MISSING}")
```

```python
from pathlib import Path
import zipfile
import io
import subprocess
import sys
import pandas as pd

# ============================================================
# INSTALAR / IMPORTAR DEPENDÊNCIAS
# ============================================================

def ensure_package(import_name, pip_name=None):
    try:
        __import__(import_name)
    except Exception:
        pkg = pip_name if pip_name is not None else import_name
        print(f"[INFO] Instalando pacote ausente: {pkg}")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])

ensure_package("scipy")
ensure_package("openpyxl")

import scipy.io

# ============================================================
# CAMINHOS
# ============================================================

ZIP_PATH = Path(r"C:\Users\Luis\Downloads\Drone-detection-dataset-master.zip")

OUT_DIR = Path(r"C:\Users\Luis\Downloads\exploracao_drone_dataset")
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_MAT_SUMMARY = OUT_DIR / "resumo_conteudo_mat_direto_zip.csv"
OUT_XLSX_SUMMARY = OUT_DIR / "resumo_xlsx_direto_zip.txt"

# ============================================================
# VERIFICAÇÃO
# ============================================================

if not ZIP_PATH.exists():
    raise FileNotFoundError(f"ZIP não encontrado: {ZIP_PATH}")

print(f"[OK] ZIP encontrado: {ZIP_PATH}")

# ============================================================
# LISTAR .MAT E .XLSX DENTRO DO ZIP
# ============================================================

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    names = z.namelist()

    mat_files = sorted([n for n in names if n.lower().endswith(".mat")])
    xlsx_files = sorted([n for n in names if n.lower().endswith(".xlsx")])

    print(f"[INFO] Arquivos .mat encontrados no ZIP : {len(mat_files)}")
    print(f"[INFO] Arquivos .xlsx encontrados no ZIP: {len(xlsx_files)}")

    print("\nPrimeiros .mat:")
    for p in mat_files[:10]:
        print(" ", p)

    print("\nArquivos .xlsx:")
    for p in xlsx_files:
        print(" ", p)

# ============================================================
# EXPLORAR CONTEÚDO DOS .MAT DIRETAMENTE DO ZIP
# ============================================================

rows = []

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    for mat_name in mat_files:
        try:
            raw = z.read(mat_name)
            data = scipy.io.loadmat(io.BytesIO(raw))

            keys = [k for k in data.keys() if not k.startswith("__")]

            row = {
                "zip_path": mat_name,
                "filename": Path(mat_name).name,
                "keys": " | ".join(keys),
                "n_keys": len(keys),
                "error": "",
            }

            for k in keys:
                value = data[k]

                row[f"{k}_type"] = type(value).__name__

                if hasattr(value, "shape"):
                    row[f"{k}_shape"] = str(value.shape)
                else:
                    row[f"{k}_shape"] = ""

                if hasattr(value, "dtype"):
                    row[f"{k}_dtype"] = str(value.dtype)
                else:
                    row[f"{k}_dtype"] = ""

            rows.append(row)

        except Exception as e:
            rows.append({
                "zip_path": mat_name,
                "filename": Path(mat_name).name,
                "keys": "",
                "n_keys": 0,
                "error": str(e),
            })

df_mat = pd.DataFrame(rows)
df_mat.to_csv(OUT_MAT_SUMMARY, index=False, encoding="utf-8-sig")

print("\n" + "=" * 80)
print("AMOSTRA DO CONTEÚDO DOS .MAT")
print("=" * 80)

if df_mat.empty:
    print("[ERRO] Nenhum .mat foi processado.")
else:
    print(df_mat.head(20).to_string(index=False))

print(f"\n[OK] Resumo dos .mat salvo em: {OUT_MAT_SUMMARY}")

# ============================================================
# EXPLORAR XLSX DIRETAMENTE DO ZIP
# ============================================================

with open(OUT_XLSX_SUMMARY, "w", encoding="utf-8") as f:
    with zipfile.ZipFile(ZIP_PATH, "r") as z:
        for xlsx_name in xlsx_files:
            f.write("=" * 80 + "\n")
            f.write(f"Arquivo: {xlsx_name}\n")
            f.write("=" * 80 + "\n\n")

            try:
                raw = z.read(xlsx_name)
                xls = pd.ExcelFile(io.BytesIO(raw))

                f.write(f"Abas encontradas: {xls.sheet_names}\n\n")

                for sheet in xls.sheet_names:
                    df_xlsx = pd.read_excel(io.BytesIO(raw), sheet_name=sheet)

                    f.write("-" * 80 + "\n")
                    f.write(f"Aba: {sheet}\n")
                    f.write(f"Shape: {df_xlsx.shape}\n")
                    f.write(f"Colunas: {list(df_xlsx.columns)}\n\n")
                    f.write(df_xlsx.head(30).to_string(index=False))
                    f.write("\n\n")

            except Exception as e:
                f.write(f"[ERRO] {e}\n\n")

print(f"[OK] Resumo do XLSX salvo em: {OUT_XLSX_SUMMARY}")
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import shutil
import re
import hashlib
from PIL import Image
from sklearn.model_selection import train_test_split

# ============================================================
# CONFIGURAÇÕES
# ============================================================

ROOT = Path(r"C:\Users\Luis\Downloads\exploracao_drone_dataset\detector_training_data")

CSV_IR = ROOT / "trainingData_IR.csv"
CSV_V  = ROOT / "trainingData_V.csv"

OUT_DIR = ROOT / "yolo_drone_detection"
OUT_IMG = OUT_DIR / "images"
OUT_LAB = OUT_DIR / "labels"

CLASSES = ["AIRPLANE", "BIRD", "DRONE", "HELICOPTER"]

CLASS_TO_ID = {
    "AIRPLANE": 0,
    "BIRD": 1,
    "DRONE": 2,
    "HELICOPTER": 3,
}

SPLITS = {
    "train": 0.70,
    "val": 0.15,
    "test": 0.15,
}

RANDOM_STATE = 42

# ============================================================
# CRIAR PASTAS
# ============================================================

for split in SPLITS:
    (OUT_IMG / split).mkdir(parents=True, exist_ok=True)
    (OUT_LAB / split).mkdir(parents=True, exist_ok=True)

OUT_DIR.mkdir(parents=True, exist_ok=True)

# ============================================================
# FUNÇÕES AUXILIARES
# ============================================================

def natural_key(col):
    """
    Ordena AIRPLANE_1, AIRPLANE_2, ..., AIRPLANE_10 corretamente.
    """
    m = re.search(r"_(\d+)$", col)
    return int(m.group(1)) if m else 999999


def extract_sample_id(path_str):
    """
    Tenta extrair o ID do vídeo original para evitar vazamento entre treino/teste.
    Exemplos esperados:
    IR_AIRPLANE_001
    V_DRONE_052
    IR_HELICOPTER_061
    """
    s = str(path_str).upper()

    pattern = r"(IR_|V_)?(AIRPLANE|BIRD|DRONE|HELICOPTER)_\d+"
    matches = re.findall(pattern, s)

    if matches:
        # re.findall retorna tuplas por causa dos grupos.
        # Vamos pegar o match completo de outro jeito.
        m = re.search(pattern, s)
        return m.group(0)

    # fallback: usa nome do arquivo sem extensão
    return Path(path_str).stem


def make_unique_image_name(image_path, sensor):
    """
    Evita colisão entre nomes de frames.
    """
    image_path = Path(image_path)
    h = hashlib.md5(str(image_path).encode("utf-8")).hexdigest()[:10]
    return f"{sensor}_{image_path.stem}_{h}{image_path.suffix.lower()}"


def read_image_size(image_path):
    with Image.open(image_path) as img:
        w, h = img.size
    return w, h


def get_bbox_groups(df, cls):
    """
    Para cada classe, encontra colunas CLASSE_1, CLASSE_2...
    e agrupa de 4 em 4: x, y, w, h.
    """
    cols = [c for c in df.columns if re.match(rf"^{cls}_\d+$", c)]
    cols = sorted(cols, key=natural_key)

    groups = []
    for i in range(0, len(cols), 4):
        g = cols[i:i+4]
        if len(g) == 4:
            groups.append(g)

    return groups


def wide_to_long(csv_path, sensor):
    """
    Converte CSV wide do MATLAB para formato longo:
    imageFilename, sensor, sample_id, class_name, class_id, x, y, w, h
    """
    df = pd.read_csv(csv_path)

    rows = []

    bbox_groups = {
        cls: get_bbox_groups(df, cls)
        for cls in CLASSES
    }

    for idx, r in df.iterrows():
        img_path = str(r["imageFilename"])
        sample_id = extract_sample_id(img_path)

        for cls in CLASSES:
            for box_id, cols in enumerate(bbox_groups[cls], start=1):
                vals = r[cols].values.astype(float)

                if np.any(pd.isna(vals)):
                    continue

                x, y, w, h = vals

                if w <= 0 or h <= 0:
                    continue

                rows.append({
                    "imageFilename": img_path,
                    "sensor": sensor,
                    "sample_id": sample_id,
                    "class_name": cls,
                    "class_id": CLASS_TO_ID[cls],
                    "box_id": box_id,
                    "x": float(x),
                    "y": float(y),
                    "width": float(w),
                    "height": float(h),
                })

    return pd.DataFrame(rows)


# ============================================================
# CONVERTER IR E VISÍVEL PARA FORMATO LONGO
# ============================================================

df_ir_long = wide_to_long(CSV_IR, sensor="IR")
df_v_long  = wide_to_long(CSV_V, sensor="V")

ann = pd.concat([df_ir_long, df_v_long], ignore_index=True)

OUT_LONG = OUT_DIR / "annotations_long.csv"
ann.to_csv(OUT_LONG, index=False, encoding="utf-8-sig")

print("[OK] annotations_long salvo em:", OUT_LONG)
print("Total de bounding boxes:", len(ann))
print(ann.head())

# ============================================================
# RESUMO POR CLASSE E SENSOR
# ============================================================

summary = (
    ann.groupby(["sensor", "class_name"])
    .agg(
        n_boxes=("class_name", "count"),
        n_images=("imageFilename", "nunique"),
        n_samples=("sample_id", "nunique"),
        mean_w=("width", "mean"),
        mean_h=("height", "mean"),
    )
    .reset_index()
)

summary["mean_w"] = summary["mean_w"].round(2)
summary["mean_h"] = summary["mean_h"].round(2)

OUT_SUMMARY = OUT_DIR / "classes_summary.csv"
summary.to_csv(OUT_SUMMARY, index=False, encoding="utf-8-sig")

print("\nResumo:")
print(summary.to_string(index=False))
print("[OK] resumo salvo em:", OUT_SUMMARY)

# ============================================================
# SPLIT POR SAMPLE_ID, NÃO POR FRAME
# ============================================================

samples = ann[["sample_id", "sensor"]].drop_duplicates().copy()
samples["group"] = samples["sensor"] + "_" + samples["sample_id"]

groups = samples["group"].unique()

train_groups, temp_groups = train_test_split(
    groups,
    train_size=SPLITS["train"],
    random_state=RANDOM_STATE,
    shuffle=True,
)

val_ratio_adjusted = SPLITS["val"] / (SPLITS["val"] + SPLITS["test"])

val_groups, test_groups = train_test_split(
    temp_groups,
    train_size=val_ratio_adjusted,
    random_state=RANDOM_STATE,
    shuffle=True,
)

group_to_split = {}

for g in train_groups:
    group_to_split[g] = "train"

for g in val_groups:
    group_to_split[g] = "val"

for g in test_groups:
    group_to_split[g] = "test"

ann["group"] = ann["sensor"] + "_" + ann["sample_id"]
ann["split"] = ann["group"].map(group_to_split)

print("\nSplit:")
print(ann.groupby("split")["imageFilename"].nunique())

# ============================================================
# GERAR DATASET YOLO
# ============================================================

image_map = {}

for img_path_str, g in ann.groupby("imageFilename"):

    img_path = Path(img_path_str)

    if not img_path.exists():
        print("[AVISO] Imagem não encontrada:", img_path)
        continue

    split = g["split"].iloc[0]
    sensor = g["sensor"].iloc[0]

    img_w, img_h = read_image_size(img_path)

    unique_name = make_unique_image_name(img_path, sensor)

    dst_img = OUT_IMG / split / unique_name
    dst_lab = OUT_LAB / split / f"{Path(unique_name).stem}.txt"

    shutil.copy2(img_path, dst_img)

    lines = []

    for _, r in g.iterrows():
        x = r["x"]
        y = r["y"]
        w = r["width"]
        h = r["height"]

        # MATLAB: x,y,w,h no canto superior esquerdo
        # YOLO: x_center, y_center, w, h normalizados
        x_center = (x + w / 2.0) / img_w
        y_center = (y + h / 2.0) / img_h
        w_norm = w / img_w
        h_norm = h / img_h

        # Limitar para [0,1]
        x_center = min(max(x_center, 0.0), 1.0)
        y_center = min(max(y_center, 0.0), 1.0)
        w_norm = min(max(w_norm, 0.0), 1.0)
        h_norm = min(max(h_norm, 0.0), 1.0)

        class_id = int(r["class_id"])

        lines.append(
            f"{class_id} {x_center:.6f} {y_center:.6f} {w_norm:.6f} {h_norm:.6f}"
        )

    with open(dst_lab, "w", encoding="utf-8") as f:
        f.write("\n".join(lines))

    image_map[str(img_path)] = str(dst_img)

# ============================================================
# SALVAR DATA.YAML
# ============================================================

yaml_text = f"""path: {OUT_DIR.as_posix()}
train: images/train
val: images/val
test: images/test

names:
  0: AIRPLANE
  1: BIRD
  2: DRONE
  3: HELICOPTER
"""

yaml_path = OUT_DIR / "data.yaml"
yaml_path.write_text(yaml_text, encoding="utf-8")

print("\n[OK] Dataset YOLO criado em:", OUT_DIR)
print("[OK] data.yaml:", yaml_path)

print("\nArquivos por split:")
for split in ["train", "val", "test"]:
    n_img = len(list((OUT_IMG / split).glob("*")))
    n_lab = len(list((OUT_LAB / split).glob("*.txt")))
    print(f"{split:5s}: imagens={n_img} labels={n_lab}")
```

```python
# ============================================================
# fusion_three_sensors_baseline.py
# ============================================================
# Objetivo:
#   Avaliar fusão de sensores para DRONE vs NON-DRONE
#
# Sensores:
#   1) IR
#   2) Vídeo visível V
#   3) Áudio
#
# Avalia:
#   - IR
#   - V
#   - Áudio
#   - IR + V
#   - IR + Áudio
#   - V + Áudio
#   - IR + V + Áudio
#
# Saídas:
#   C:\Users\Luis\Downloads\exploracao_drone_dataset\fusion_3sensors_baseline
# ============================================================

from pathlib import Path
import re
import sys
import subprocess
import warnings
import itertools
import json
import random

warnings.filterwarnings("ignore")


# ============================================================
# INSTALAÇÃO / IMPORTS
# ============================================================

def ensure_package(import_name, pip_name=None):
    try:
        __import__(import_name)
    except Exception:
        pkg = pip_name if pip_name is not None else import_name
        print(f"[INFO] Instalando pacote ausente: {pkg}")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])


ensure_package("numpy")
ensure_package("pandas")
ensure_package("PIL", "pillow")
ensure_package("sklearn", "scikit-learn")
ensure_package("librosa")
ensure_package("soundfile")

import numpy as np
import pandas as pd
from PIL import Image

import librosa

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
)


# ============================================================
# CONFIGURAÇÕES
# ============================================================

DATA_ROOT = Path(
    r"C:\Users\Luis\Downloads\DroneDetectionDataset\Drone-detection-dataset-master\Data"
)

DETECT_ROOT = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\detector_training_data"
)

CSV_IR = DETECT_ROOT / "trainingData_IR.csv"
CSV_V = DETECT_ROOT / "trainingData_V.csv"

AUDIO_DIR = DATA_ROOT / "Audio"

OUT_DIR = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\fusion_3sensors_baseline"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
MAX_FRAMES_PER_SAMPLE = 60
IMG_SIZE = 32
AUDIO_SR = 16000

CLASSES_VIDEO = ["AIRPLANE", "BIRD", "DRONE", "HELICOPTER"]
CLASSES_AUDIO = ["BACKGROUND", "DRONE", "HELICOPTER"]


# ============================================================
# FUNÇÕES GERAIS
# ============================================================

def normalize_sample_key(text):
    """
    Extrai chaves como:
      IR_DRONE_001 -> DRONE_001
      V_HELICOPTER_061 -> HELICOPTER_061
      DRONE_001.wav -> DRONE_001
    """

    s = str(text).upper().replace("\\", "/")

    m = re.search(
        r"(?:IR_|V_)?(AIRPLANE|BIRD|DRONE|HELICOPTER|BACKGROUND)_(\d+)",
        s
    )

    if m:
        cls = m.group(1)
        num = int(m.group(2))
        return f"{cls}_{num:03d}"

    return Path(str(text)).stem.upper()


def class_from_sample_key(sample_key):
    return str(sample_key).split("_")[0].upper()


def binary_label_from_class(class_name):
    """
    DRONE = 1
    Todo o resto = 0
    """
    return 1 if str(class_name).upper() == "DRONE" else 0


def safe_train_test_split(df, test_size=0.30):
    """
    Split estratificado por amostra.
    """
    y = df["y"].values

    if len(np.unique(y)) < 2:
        raise ValueError("Apenas uma classe disponível.")

    counts = pd.Series(y).value_counts()

    if counts.min() < 2:
        raise ValueError("Poucas amostras em pelo menos uma classe.")

    return train_test_split(
        df,
        test_size=test_size,
        random_state=RANDOM_STATE,
        stratify=y,
        shuffle=True,
    )


# ============================================================
# FEATURES DE IMAGEM
# ============================================================

def image_features(image_path):
    """
    Extrai features simples de imagem:
      - imagem cinza 32x32 achatada
      - histograma de intensidade
      - estatísticas
      - gradiente simples
    """

    image_path = Path(image_path)

    with Image.open(image_path) as img:
        img = img.convert("L")
        img = img.resize((IMG_SIZE, IMG_SIZE))
        arr = np.asarray(img).astype(np.float32) / 255.0

    flat = arr.reshape(-1)

    hist, _ = np.histogram(arr, bins=16, range=(0.0, 1.0), density=True)

    gy, gx = np.gradient(arr)
    grad_mag = np.sqrt(gx**2 + gy**2)
    grad_hist, _ = np.histogram(
        grad_mag,
        bins=16,
        range=(0.0, max(float(grad_mag.max()), 1e-6)),
        density=True
    )

    stats = np.array([
        arr.mean(),
        arr.std(),
        arr.min(),
        arr.max(),
        grad_mag.mean(),
        grad_mag.std(),
    ], dtype=np.float32)

    feat = np.concatenate([flat, hist, grad_hist, stats]).astype(np.float32)

    feat = np.nan_to_num(feat, nan=0.0, posinf=0.0, neginf=0.0)

    return feat


def build_video_feature_table(csv_path, sensor_name):
    """
    Constrói uma tabela sample-level para IR ou V.
    Cada sample recebe a média e o desvio-padrão das features dos frames.
    """

    csv_path = Path(csv_path)

    if not csv_path.exists():
        raise FileNotFoundError(csv_path)

    df = pd.read_csv(csv_path)

    if "imageFilename" not in df.columns:
        raise ValueError(f"CSV sem coluna imageFilename: {csv_path}")

    df["sample_key"] = df["imageFilename"].apply(normalize_sample_key)
    df["class_name"] = df["sample_key"].apply(class_from_sample_key)
    df["y"] = df["class_name"].apply(binary_label_from_class)

    rows = []

    grouped = df.groupby("sample_key")

    print(f"\n[INFO] Extraindo features de vídeo: {sensor_name}")
    print(f"[INFO] Samples encontrados: {len(grouped)}")

    rng = np.random.default_rng(RANDOM_STATE)

    for sample_key, g in grouped:

        img_paths = list(g["imageFilename"].dropna().unique())

        existing = [p for p in img_paths if Path(p).exists()]

        if len(existing) == 0:
            print(f"[AVISO] Nenhuma imagem encontrada para {sample_key}")
            continue

        if len(existing) > MAX_FRAMES_PER_SAMPLE:
            idx = rng.choice(len(existing), size=MAX_FRAMES_PER_SAMPLE, replace=False)
            existing = [existing[i] for i in idx]

        feats = []

        for p in existing:
            try:
                feats.append(image_features(p))
            except Exception as e:
                print(f"[AVISO] Erro em imagem {p}: {e}")

        if len(feats) == 0:
            continue

        feats = np.vstack(feats)

        feat_mean = feats.mean(axis=0)
        feat_std = feats.std(axis=0)

        sample_feat = np.concatenate([feat_mean, feat_std])

        class_name = class_from_sample_key(sample_key)
        y = binary_label_from_class(class_name)

        row = {
            "sample_key": sample_key,
            "class_name": class_name,
            "y": y,
            "n_frames_used": len(existing),
        }

        for i, val in enumerate(sample_feat):
            row[f"{sensor_name}_f{i:04d}"] = float(val)

        rows.append(row)

    out = pd.DataFrame(rows)

    print(f"[OK] {sensor_name}: {out.shape[0]} samples, {out.shape[1]} colunas")

    return out


# ============================================================
# FEATURES DE ÁUDIO
# ============================================================

def audio_features(wav_path):
    """
    Extrai features de áudio:
      - log-mel mean/std
      - MFCC mean/std
      - RMS, ZCR, centroid, bandwidth, rolloff
    """

    wav_path = Path(wav_path)

    y, sr = librosa.load(wav_path, sr=AUDIO_SR, mono=True)

    if len(y) < sr // 2:
        y = np.pad(y, (0, sr // 2 - len(y)))

    mel = librosa.feature.melspectrogram(
        y=y,
        sr=sr,
        n_fft=1024,
        hop_length=256,
        n_mels=64,
        fmin=20,
        fmax=sr // 2,
        power=2.0,
    )

    logmel = librosa.power_to_db(mel + 1e-12, ref=np.max)

    mfcc = librosa.feature.mfcc(
        S=logmel,
        n_mfcc=20
    )

    rms = librosa.feature.rms(y=y, frame_length=1024, hop_length=256)
    zcr = librosa.feature.zero_crossing_rate(y, frame_length=1024, hop_length=256)
    centroid = librosa.feature.spectral_centroid(y=y, sr=sr, n_fft=1024, hop_length=256)
    bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr, n_fft=1024, hop_length=256)
    rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr, n_fft=1024, hop_length=256)

    parts = []

    for mat in [logmel, mfcc, rms, zcr, centroid, bandwidth, rolloff]:
        parts.append(mat.mean(axis=1))
        parts.append(mat.std(axis=1))

    feat = np.concatenate(parts).astype(np.float32)
    feat = np.nan_to_num(feat, nan=0.0, posinf=0.0, neginf=0.0)

    return feat


def build_audio_feature_table(audio_dir):
    audio_dir = Path(audio_dir)

    if not audio_dir.exists():
        raise FileNotFoundError(audio_dir)

    wavs = sorted(audio_dir.glob("*.wav"))

    rows = []

    print("\n[INFO] Extraindo features de áudio")
    print(f"[INFO] WAVs encontrados: {len(wavs)}")

    for wav_path in wavs:
        sample_key = normalize_sample_key(wav_path.name)
        class_name = class_from_sample_key(sample_key)
        y = binary_label_from_class(class_name)

        try:
            feat = audio_features(wav_path)
        except Exception as e:
            print(f"[AVISO] Erro em áudio {wav_path}: {e}")
            continue

        row = {
            "sample_key": sample_key,
            "class_name": class_name,
            "y": y,
            "audio_path": str(wav_path),
        }

        for i, val in enumerate(feat):
            row[f"audio_f{i:04d}"] = float(val)

        rows.append(row)

    out = pd.DataFrame(rows)

    print(f"[OK] audio: {out.shape[0]} samples, {out.shape[1]} colunas")

    return out


# ============================================================
# COMBINAR MODALIDADES
# ============================================================

def feature_columns(df):
    return [
        c for c in df.columns
        if re.search(r"_f\d+$", c)
    ]


def prepare_combination(tables, combo):
    """
    Faz inner join por sample_key.
    Só entram amostras que existem em todos os sensores da combinação.
    """

    dfs = []

    for name in combo:
        df = tables[name].copy()

        keep = ["sample_key", "class_name", "y"] + feature_columns(df)
        df = df[keep]

        # evita colunas duplicadas de metadata
        df = df.rename(columns={
            "class_name": f"class_name_{name}",
            "y": f"y_{name}",
        })

        dfs.append(df)

    merged = dfs[0]

    for df in dfs[1:]:
        merged = merged.merge(df, on="sample_key", how="inner")

    y_cols = [c for c in merged.columns if c.startswith("y_")]

    merged["y"] = merged[y_cols[0]]

    # Confere se todos os sensores concordam no rótulo
    for yc in y_cols[1:]:
        bad = merged[merged[yc] != merged["y"]]
        if len(bad) > 0:
            print(f"[AVISO] Rótulos conflitantes em {combo}: {len(bad)} linhas")

    class_cols = [c for c in merged.columns if c.startswith("class_name_")]
    merged["class_name"] = merged[class_cols[0]]

    X_cols = feature_columns(merged)

    return merged, X_cols


# ============================================================
# TREINAR E AVALIAR
# ============================================================

def evaluate_model(df, X_cols, combo_name, model_name, model):
    """
    Avalia uma combinação de sensores.
    """

    if len(df) < 10:
        print(f"[SKIP] {combo_name}: poucas amostras ({len(df)})")
        return None, None

    if df["y"].nunique() < 2:
        print(f"[SKIP] {combo_name}: apenas uma classe")
        return None, None

    try:
        train_df, test_df = safe_train_test_split(df, test_size=0.30)
    except Exception as e:
        print(f"[SKIP] {combo_name}: {e}")
        return None, None

    X_train = train_df[X_cols].values
    y_train = train_df["y"].values

    X_test = test_df[X_cols].values
    y_test = test_df["y"].values

    pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("clf", model),
    ])

    pipe.fit(X_train, y_train)

    y_pred = pipe.predict(X_test)

    if hasattr(pipe, "predict_proba"):
        y_score = pipe.predict_proba(X_test)[:, 1]
    else:
        y_score = y_pred.astype(float)

    metrics = {
        "combo": combo_name,
        "model": model_name,
        "n_samples": int(len(df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(X_cols)),
        "positive_rate": float(df["y"].mean()),
        "accuracy": float(accuracy_score(y_test, y_pred)),
        "balanced_accuracy": float(balanced_accuracy_score(y_test, y_pred)),
        "precision": float(precision_score(y_test, y_pred, zero_division=0)),
        "recall": float(recall_score(y_test, y_pred, zero_division=0)),
        "f1": float(f1_score(y_test, y_pred, zero_division=0)),
    }

    try:
        metrics["roc_auc"] = float(roc_auc_score(y_test, y_score))
    except Exception:
        metrics["roc_auc"] = np.nan

    cm = confusion_matrix(y_test, y_pred)

    pred_df = test_df[["sample_key", "class_name", "y"]].copy()
    pred_df["y_pred"] = y_pred
    pred_df["score_drone"] = y_score
    pred_df["combo"] = combo_name
    pred_df["model"] = model_name

    print("\n" + "=" * 80)
    print(f"{combo_name} | {model_name}")
    print("=" * 80)
    print(json.dumps(metrics, indent=2))
    print("Confusion matrix:")
    print(cm)
    print(classification_report(
        y_test,
        y_pred,
        target_names=["NON_DRONE", "DRONE"],
        zero_division=0
    ))

    return metrics, pred_df


# ============================================================
# MAIN
# ============================================================

def main():

    random.seed(RANDOM_STATE)
    np.random.seed(RANDOM_STATE)

    print("=" * 80)
    print("FUSÃO DE 3 SENSORES — BASELINE SAMPLE-LEVEL")
    print("=" * 80)

    tables = {}

    tables["IR"] = build_video_feature_table(CSV_IR, sensor_name="IR")
    tables["V"] = build_video_feature_table(CSV_V, sensor_name="V")
    tables["AUDIO"] = build_audio_feature_table(AUDIO_DIR)

    # Salvar features unimodais
    for name, df in tables.items():
        out = OUT_DIR / f"features_{name}.csv"
        df.to_csv(out, index=False, encoding="utf-8-sig")
        print(f"[OK] Features {name} salvas em: {out}")

    combos = [
        ("IR",),
        ("V",),
        ("AUDIO",),
        ("IR", "V"),
        ("IR", "AUDIO"),
        ("V", "AUDIO"),
        ("IR", "V", "AUDIO"),
    ]

    models = {
        "logreg": LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            solver="lbfgs"
        ),
        "rf": RandomForestClassifier(
            n_estimators=400,
            random_state=RANDOM_STATE,
            class_weight="balanced",
            n_jobs=-1,
            max_depth=None,
        ),
    }

    all_metrics = []
    all_preds = []

    combo_sizes = []

    for combo in combos:
        combo_name = "+".join(combo)

        merged, X_cols = prepare_combination(tables, combo)

        combo_sizes.append({
            "combo": combo_name,
            "n_samples": len(merged),
            "n_drone": int(merged["y"].sum()) if len(merged) else 0,
            "n_non_drone": int((merged["y"] == 0).sum()) if len(merged) else 0,
            "n_features": len(X_cols),
        })

        merged_out = OUT_DIR / f"merged_{combo_name.replace('+','_')}.csv"
        merged[["sample_key", "class_name", "y"]].to_csv(
            merged_out,
            index=False,
            encoding="utf-8-sig"
        )

        print("\n" + "-" * 80)
        print(f"Combinação: {combo_name}")
        print(f"Amostras: {len(merged)}")
        print(f"Features: {len(X_cols)}")
        if len(merged):
            print(merged["y"].value_counts().rename(index={0: "NON_DRONE", 1: "DRONE"}))
        print("-" * 80)

        for model_name, model in models.items():
            metrics, pred_df = evaluate_model(
                merged,
                X_cols,
                combo_name,
                model_name,
                model
            )

            if metrics is not None:
                all_metrics.append(metrics)

            if pred_df is not None:
                all_preds.append(pred_df)

    sizes_df = pd.DataFrame(combo_sizes)
    sizes_path = OUT_DIR / "combination_sizes.csv"
    sizes_df.to_csv(sizes_path, index=False, encoding="utf-8-sig")

    metrics_df = pd.DataFrame(all_metrics)
    metrics_path = OUT_DIR / "fusion_metrics.csv"
    metrics_df.to_csv(metrics_path, index=False, encoding="utf-8-sig")

    if all_preds:
        preds_df = pd.concat(all_preds, ignore_index=True)
        preds_path = OUT_DIR / "fusion_predictions.csv"
        preds_df.to_csv(preds_path, index=False, encoding="utf-8-sig")
    else:
        preds_path = None

    print("\n" + "=" * 80)
    print("RESUMO FINAL")
    print("=" * 80)

    print("\nTamanho das combinações:")
    print(sizes_df.to_string(index=False))

    print("\nMétricas:")
    if not metrics_df.empty:
        print(
            metrics_df.sort_values(
                ["balanced_accuracy", "f1", "roc_auc"],
                ascending=False
            ).to_string(index=False)
        )
    else:
        print("[AVISO] Nenhuma métrica gerada.")

    print("\nArquivos salvos:")
    print(f"[OK] {sizes_path}")
    print(f"[OK] {metrics_path}")

    if preds_path is not None:
        print(f"[OK] {preds_path}")

    print(f"\n[OK] Diretório de saída:\n{OUT_DIR}")


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# multiclass_fusion_with_audio_FIXED.py
# ============================================================
# Classificação multiclasse com fusão de sensores:
#
# Sensores:
#   IR      -> vídeo infravermelho
#   V       -> vídeo visível
#   AUDIO   -> áudio
#
# Classes globais:
#   AIRPLANE
#   BACKGROUND
#   BIRD
#   DRONE
#   HELICOPTER
#
# Avaliações:
#   - unimodal
#   - early fusion por concatenação
#   - late fusion por média
#   - late fusion ponderada por OOF
#   - late fusion por stacking
#
# Saídas:
#   C:\Users\Luis\Downloads\exploracao_drone_dataset\multiclass_fusion_audio_results_fixed
# ============================================================

from pathlib import Path
import sys
import subprocess
import warnings
import re
import json
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ============================================================
# INSTALAÇÃO / IMPORTS
# ============================================================

def ensure_package(import_name, pip_name=None, optional=False):
    try:
        return __import__(import_name)
    except Exception:
        if optional:
            return None
        pkg = pip_name if pip_name is not None else import_name
        print(f"[INFO] Instalando pacote ausente: {pkg}")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])
        return __import__(import_name)


ensure_package("numpy")
ensure_package("pandas")
ensure_package("sklearn", "scikit-learn")
ensure_package("matplotlib")

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
)
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix,
)
import matplotlib.pyplot as plt


# ============================================================
# CONFIGURAÇÕES
# ============================================================

FEATURE_ROOT = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\fusion_3sensors_baseline"
)

FEATURE_FILES = {
    "IR": FEATURE_ROOT / "features_IR.csv",
    "V": FEATURE_ROOT / "features_V.csv",
    "AUDIO": FEATURE_ROOT / "features_AUDIO.csv",
}

OUT_DIR = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\multiclass_fusion_audio_results_fixed"
)
OUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42

# Para teste rápido, use 3.
# Para resultado final, use 10 ou 20.
N_REPEATS = 10

TEST_SIZE = 0.30
MAX_CV_FOLDS = 5

# Se quiser instalar automaticamente XGBoost/LightGBM/CatBoost, coloque True.
INSTALL_OPTIONAL_SOTA = False

SENSOR_LIST = ["IR", "V", "AUDIO"]

COMBOS = [
    ("IR",),
    ("V",),
    ("AUDIO",),
    ("IR", "V"),
    ("IR", "AUDIO"),
    ("V", "AUDIO"),
    ("IR", "V", "AUDIO"),
]


# ============================================================
# FEATURES / CLASSES
# ============================================================

def normalize_class_name(x):
    return str(x).upper().strip()


def get_feature_cols(df, sensor=None):
    """
    Retorna colunas de features.
    Aceita:
      IR_f0000
      V_f0000
      audio_f0000
      AUDIO_f0000
    """
    if sensor is None:
        return [
            c for c in df.columns
            if re.search(r"_f\d+$", c, flags=re.IGNORECASE)
        ]

    prefix = f"{sensor}_f".upper()
    return [
        c for c in df.columns
        if c.upper().startswith(prefix)
    ]


def standardize_feature_columns(df, sensor):
    """
    Padroniza nomes de features:
      audio_f0000 -> AUDIO_f0000
      IR_f0000    -> IR_f0000
      V_f0000     -> V_f0000
    """
    rename = {}

    for c in df.columns:
        if re.search(r"_f\d+$", c, flags=re.IGNORECASE):
            m = re.search(r"_f(\d+)$", c, flags=re.IGNORECASE)
            if m:
                num = int(m.group(1))
                # aceita apenas colunas do sensor
                if c.upper().startswith(f"{sensor}_F") or (
                    sensor == "AUDIO" and c.upper().startswith("AUDIO_F")
                ):
                    rename[c] = f"{sensor}_f{num:04d}"

    if rename:
        df = df.rename(columns=rename)

    return df


def force_numeric_features(df, feature_cols):
    """
    Converte features para float, substituindo texto inválido por NaN.
    """
    df = df.copy()

    for c in feature_cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df


def load_sensor_tables():
    tables = {}

    for sensor, path in FEATURE_FILES.items():

        if not path.exists():
            raise FileNotFoundError(f"Arquivo não encontrado: {path}")

        df = pd.read_csv(path)

        required = {"sample_key", "class_name"}
        missing = required - set(df.columns)

        if missing:
            raise ValueError(f"{path} sem colunas obrigatórias: {missing}")

        df["sample_key"] = df["sample_key"].astype(str)
        df["class_name"] = df["class_name"].apply(normalize_class_name)

        # Corrige audio_f0000 -> AUDIO_f0000
        df = standardize_feature_columns(df, sensor)

        fcols = get_feature_cols(df, sensor)

        if len(fcols) == 0:
            raise ValueError(f"Nenhuma feature encontrada para sensor {sensor}")

        df = force_numeric_features(df, fcols)

        # Mantém só sample_key, class_name e features
        df = df[["sample_key", "class_name"] + fcols].drop_duplicates("sample_key")

        tables[sensor] = df

        print(
            f"[OK] {sensor}: {df.shape[0]} amostras, "
            f"{len(fcols)} features, classes={sorted(df['class_name'].unique())}"
        )

    return tables


def print_class_availability(tables):
    rows = []

    for sensor, df in tables.items():
        for cls, g in df.groupby("class_name"):
            rows.append({
                "sensor": sensor,
                "class_name": cls,
                "n_samples": int(g["sample_key"].nunique()),
            })

    out = pd.DataFrame(rows).sort_values(["sensor", "class_name"])
    path = OUT_DIR / "class_availability_by_sensor.csv"
    out.to_csv(path, index=False, encoding="utf-8-sig")

    print("\n" + "=" * 80)
    print("CLASSES DISPONÍVEIS POR SENSOR")
    print("=" * 80)
    print(out.to_string(index=False))
    print(f"[OK] Salvo em: {path}")

    return out


def build_union_dataset(tables, combo):
    """
    União de amostras:
    entra qualquer sample_key que exista em pelo menos um sensor da combinação.

    Features de sensores ausentes ficam NaN.
    Adiciona has_IR, has_V, has_AUDIO.
    """

    meta_rows = []

    for sensor in combo:
        df = tables[sensor]
        for _, r in df[["sample_key", "class_name"]].iterrows():
            meta_rows.append({
                "sample_key": r["sample_key"],
                "class_name": r["class_name"],
                "sensor": sensor,
            })

    meta = pd.DataFrame(meta_rows)

    # Classe por sample_key
    # Se aparecer em mais de um sensor, usa a moda.
    class_by_sample = (
        meta.groupby("sample_key")["class_name"]
        .agg(lambda x: x.value_counts().index[0])
        .reset_index()
    )

    merged = class_by_sample.copy()

    for sensor in combo:
        df = tables[sensor].copy()
        fcols = get_feature_cols(df, sensor)

        sensor_df = df[["sample_key"] + fcols].drop_duplicates("sample_key")

        merged = merged.merge(sensor_df, on="sample_key", how="left")

        merged[f"has_{sensor}"] = (
            merged[fcols].notna().any(axis=1).astype(int)
        )

    return merged


def get_combo_feature_cols(df, combo):
    cols = []

    for sensor in combo:
        cols.extend(get_feature_cols(df, sensor))

    for sensor in combo:
        has_col = f"has_{sensor}"
        if has_col in df.columns:
            cols.append(has_col)

    return cols


# ============================================================
# MODELOS
# ============================================================

def make_pipeline(estimator, scale=True):
    steps = [("imputer", SimpleImputer(strategy="median"))]

    if scale:
        steps.append(("scaler", StandardScaler()))

    steps.append(("clf", estimator))
    return Pipeline(steps)


def model_zoo(random_state=42):
    models = {}

    models["dummy_prior"] = lambda: make_pipeline(
        DummyClassifier(strategy="prior", random_state=random_state),
        scale=False,
    )

    # Corrigido: sem multi_class
    models["logreg_l2"] = lambda: make_pipeline(
        LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            solver="lbfgs",
            random_state=random_state,
        ),
        scale=True,
    )

    models["linear_svm"] = lambda: make_pipeline(
        SVC(
            kernel="linear",
            probability=True,
            class_weight="balanced",
            random_state=random_state,
        ),
        scale=True,
    )

    models["random_forest"] = lambda: make_pipeline(
        RandomForestClassifier(
            n_estimators=500,
            class_weight="balanced",
            random_state=random_state,
            n_jobs=-1,
        ),
        scale=False,
    )

    models["extra_trees"] = lambda: make_pipeline(
        ExtraTreesClassifier(
            n_estimators=700,
            class_weight="balanced",
            random_state=random_state,
            n_jobs=-1,
        ),
        scale=False,
    )

    models["hist_gb"] = lambda: make_pipeline(
        HistGradientBoostingClassifier(
            learning_rate=0.05,
            max_iter=300,
            l2_regularization=0.01,
            random_state=random_state,
        ),
        scale=False,
    )

    models["mlp"] = lambda: make_pipeline(
        MLPClassifier(
            hidden_layer_sizes=(256, 64),
            activation="relu",
            alpha=1e-4,
            learning_rate_init=1e-3,
            max_iter=800,
            early_stopping=True,
            random_state=random_state,
        ),
        scale=True,
    )

    if INSTALL_OPTIONAL_SOTA:
        for pkg in ["xgboost", "lightgbm", "catboost"]:
            try:
                __import__(pkg)
            except Exception:
                print(f"[INFO] Instalando pacote opcional: {pkg}")
                subprocess.check_call([
                    sys.executable, "-m", "pip", "install", "-q", pkg
                ])

    try:
        from xgboost import XGBClassifier

        models["xgboost"] = lambda: make_pipeline(
            XGBClassifier(
                n_estimators=500,
                max_depth=3,
                learning_rate=0.03,
                subsample=0.90,
                colsample_bytree=0.90,
                objective="multi:softprob",
                eval_metric="mlogloss",
                random_state=random_state,
                n_jobs=-1,
            ),
            scale=False,
        )
    except Exception:
        print("[INFO] XGBoost não disponível.")

    try:
        from lightgbm import LGBMClassifier

        models["lightgbm"] = lambda: make_pipeline(
            LGBMClassifier(
                n_estimators=700,
                learning_rate=0.03,
                num_leaves=31,
                subsample=0.90,
                colsample_bytree=0.90,
                class_weight="balanced",
                random_state=random_state,
                n_jobs=-1,
                verbose=-1,
            ),
            scale=False,
        )
    except Exception:
        print("[INFO] LightGBM não disponível.")

    try:
        from catboost import CatBoostClassifier

        models["catboost"] = lambda: make_pipeline(
            CatBoostClassifier(
                iterations=700,
                depth=4,
                learning_rate=0.03,
                loss_function="MultiClass",
                random_seed=random_state,
                verbose=False,
                auto_class_weights="Balanced",
            ),
            scale=False,
        )
    except Exception:
        print("[INFO] CatBoost não disponível.")

    return models


# ============================================================
# TREINO COM RÓTULOS LOCAIS
# ============================================================

def class_names_to_global_ids(y_names, global_class_to_id):
    return np.array([global_class_to_id[x] for x in y_names], dtype=int)


def local_encode(y_global):
    """
    Converte ids globais para ids locais 0..K-1.
    Necessário para XGBoost e outros modelos.
    """
    local_classes = sorted(np.unique(y_global).tolist())
    global_to_local = {g: i for i, g in enumerate(local_classes)}

    y_local = np.array([global_to_local[g] for g in y_global], dtype=int)

    return y_local, local_classes


def predict_proba_global_from_local_model(model, X, local_classes, n_global_classes):
    """
    O modelo foi treinado com labels locais.
    Aqui convertemos as probabilidades para o espaço global.
    """
    if hasattr(model, "predict_proba"):
        p_local = model.predict_proba(X)
    else:
        y_pred_local = model.predict(X)
        p_local = np.zeros((len(y_pred_local), len(local_classes)), dtype=float)
        for i, yy in enumerate(y_pred_local):
            p_local[i, int(yy)] = 1.0

    out = np.zeros((p_local.shape[0], n_global_classes), dtype=float)

    model_classes = model.named_steps["clf"].classes_

    for j, local_label in enumerate(model_classes):
        local_label = int(local_label)
        global_id = local_classes[local_label]
        out[:, global_id] = p_local[:, j]

    row_sum = out.sum(axis=1, keepdims=True)
    row_sum[row_sum <= 1e-12] = 1.0
    out = out / row_sum

    return out


def fit_model_local_labels(train_df, x_cols, model_factory, global_class_to_id):
    X_train = train_df[x_cols].apply(pd.to_numeric, errors="coerce").values.astype(float)

    y_global = class_names_to_global_ids(
        train_df["class_name"].values,
        global_class_to_id,
    )

    y_local, local_classes = local_encode(y_global)

    model = model_factory()
    model.fit(X_train, y_local)

    return model, local_classes


def predict_model_global(model, local_classes, test_df, x_cols, n_global_classes):
    X_test = test_df[x_cols].apply(pd.to_numeric, errors="coerce").values.astype(float)

    return predict_proba_global_from_local_model(
        model,
        X_test,
        local_classes,
        n_global_classes,
    )


# ============================================================
# MÉTRICAS
# ============================================================

def metrics_multiclass(y_true_names, y_pred_names, labels):
    return {
        "accuracy": float(accuracy_score(y_true_names, y_pred_names)),
        "balanced_accuracy": float(balanced_accuracy_score(y_true_names, y_pred_names)),
        "macro_precision": float(
            precision_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_recall": float(
            recall_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_f1": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="macro",
                zero_division=0,
            )
        ),
        "weighted_f1": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="weighted",
                zero_division=0,
            )
        ),
    }


def safe_stratified_split(df, random_state):
    counts = df["class_name"].value_counts()

    # Remove classes com menos de 2 amostras
    valid = counts[counts >= 2].index.tolist()
    df = df[df["class_name"].isin(valid)].copy()

    if df["class_name"].nunique() < 2:
        raise ValueError("Menos de duas classes com amostras suficientes.")

    train_df, test_df = train_test_split(
        df,
        test_size=TEST_SIZE,
        random_state=random_state,
        stratify=df["class_name"],
        shuffle=True,
    )

    return train_df.reset_index(drop=True), test_df.reset_index(drop=True)


# ============================================================
# AVALIAÇÃO DIRETA
# ============================================================

def evaluate_direct(
    train_df,
    test_df,
    x_cols,
    model_name,
    model_factory,
    combo_name,
    mode,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    model, local_classes = fit_model_local_labels(
        train_df,
        x_cols,
        model_factory,
        global_class_to_id,
    )

    p_global = predict_model_global(
        model,
        local_classes,
        test_df,
        x_cols,
        n_global,
    )

    pred_ids = np.argmax(p_global, axis=1)
    y_pred = np.array([global_classes[i] for i in pred_ids])
    y_true = test_df["class_name"].values

    m = metrics_multiclass(y_true, y_pred, global_classes)

    m.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": mode,
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(x_cols)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    preds = test_df[["sample_key", "class_name"]].copy()
    preds["repeat"] = repeat
    preds["combo"] = combo_name
    preds["mode"] = mode
    preds["model"] = model_name
    preds["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        preds[f"prob_{cls}"] = p_global[:, i]

    return m, preds


# ============================================================
# LATE FUSION
# ============================================================

def oof_sensor_predictions(
    train_df,
    test_df,
    sensor,
    model_factory,
    global_classes,
    global_class_to_id,
    random_state,
):
    fcols = get_feature_cols(train_df, sensor)
    has_col = f"has_{sensor}"

    n_global = len(global_classes)

    oof = np.zeros((len(train_df), n_global), dtype=float)
    test_prob = np.zeros((len(test_df), n_global), dtype=float)

    train_idx = np.where(train_df[has_col].values == 1)[0]
    test_idx = np.where(test_df[has_col].values == 1)[0]

    if len(train_idx) < 4:
        return oof, test_prob, 0.0

    tr_sensor = train_df.iloc[train_idx].copy()

    if tr_sensor["class_name"].nunique() < 2:
        return oof, test_prob, 0.0

    y_names = tr_sensor["class_name"].values
    y_global = class_names_to_global_ids(y_names, global_class_to_id)

    min_count = pd.Series(y_names).value_counts().min()
    n_splits = min(MAX_CV_FOLDS, int(min_count))

    if n_splits < 2:
        return oof, test_prob, 0.0

    X_sensor = tr_sensor[fcols].apply(pd.to_numeric, errors="coerce").values.astype(float)

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state,
    )

    for tr, va in skf.split(X_sensor, y_global):
        fold_df_train = tr_sensor.iloc[tr].copy()
        fold_df_val = tr_sensor.iloc[va].copy()

        model, local_classes = fit_model_local_labels(
            fold_df_train,
            fcols,
            model_factory,
            global_class_to_id,
        )

        p_val = predict_model_global(
            model,
            local_classes,
            fold_df_val,
            fcols,
            n_global,
        )

        original_val_idx = train_idx[va]
        oof[original_val_idx, :] = p_val

    # Modelo final
    model, local_classes = fit_model_local_labels(
        tr_sensor,
        fcols,
        model_factory,
        global_class_to_id,
    )

    if len(test_idx) > 0:
        te_sensor = test_df.iloc[test_idx].copy()

        p_test = predict_model_global(
            model,
            local_classes,
            te_sensor,
            fcols,
            n_global,
        )

        test_prob[test_idx, :] = p_test

    # Peso OOF por balanced accuracy
    available_oof = oof[train_idx]
    pred_ids = np.argmax(available_oof, axis=1)
    y_pred = np.array([global_classes[i] for i in pred_ids])
    y_true = tr_sensor["class_name"].values

    try:
        weight = balanced_accuracy_score(y_true, y_pred)
    except Exception:
        weight = 0.0

    return oof, test_prob, float(weight)


def normalize_rows(p):
    p = np.asarray(p, dtype=float)
    s = p.sum(axis=1, keepdims=True)
    s[s <= 1e-12] = 1.0
    return p / s


def evaluate_late_fusion(
    train_df,
    test_df,
    combo,
    combo_name,
    model_name,
    model_factory,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    sensor_test = {}
    sensor_weights = {}

    for sensor in combo:
        _, ptest, weight = oof_sensor_predictions(
            train_df,
            test_df,
            sensor,
            model_factory,
            global_classes,
            global_class_to_id,
            random_state=RANDOM_STATE + repeat,
        )

        sensor_test[sensor] = ptest
        sensor_weights[sensor] = weight

    y_true = test_df["class_name"].values

    # ----------------------------
    # Média simples
    # ----------------------------
    p_sum = np.zeros((len(test_df), n_global), dtype=float)
    count = np.zeros((len(test_df), 1), dtype=float)

    for sensor in combo:
        has = test_df[f"has_{sensor}"].values.reshape(-1, 1).astype(float)
        p_sum += sensor_test[sensor] * has
        count += has

    count[count <= 1e-12] = 1.0
    p_mean = normalize_rows(p_sum / count)

    y_pred = np.array([global_classes[i] for i in np.argmax(p_mean, axis=1)])

    m_mean = metrics_multiclass(y_true, y_pred, global_classes)
    m_mean.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_mean",
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(combo)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    pred_mean = test_df[["sample_key", "class_name"]].copy()
    pred_mean["repeat"] = repeat
    pred_mean["combo"] = combo_name
    pred_mean["mode"] = "late_mean"
    pred_mean["model"] = model_name
    pred_mean["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        pred_mean[f"prob_{cls}"] = p_mean[:, i]

    # ----------------------------
    # Média ponderada
    # ----------------------------
    p_wsum = np.zeros((len(test_df), n_global), dtype=float)
    w_sum = np.zeros((len(test_df), 1), dtype=float)

    for sensor in combo:
        has = test_df[f"has_{sensor}"].values.reshape(-1, 1).astype(float)
        w = max(sensor_weights.get(sensor, 0.0), 1e-6)

        p_wsum += sensor_test[sensor] * has * w
        w_sum += has * w

    w_sum[w_sum <= 1e-12] = 1.0
    p_weighted = normalize_rows(p_wsum / w_sum)

    y_pred_w = np.array([global_classes[i] for i in np.argmax(p_weighted, axis=1)])

    m_weighted = metrics_multiclass(y_true, y_pred_w, global_classes)
    m_weighted.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_weighted_oof",
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(combo)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    pred_weighted = test_df[["sample_key", "class_name"]].copy()
    pred_weighted["repeat"] = repeat
    pred_weighted["combo"] = combo_name
    pred_weighted["mode"] = "late_weighted_oof"
    pred_weighted["model"] = model_name
    pred_weighted["pred_class"] = y_pred_w

    for i, cls in enumerate(global_classes):
        pred_weighted[f"prob_{cls}"] = p_weighted[:, i]

    return [m_mean, m_weighted], [pred_mean, pred_weighted]


def evaluate_late_stacking(
    train_df,
    test_df,
    combo,
    combo_name,
    base_models,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    train_parts = []
    test_parts = []

    for model_name, factory in base_models.items():
        for sensor in combo:
            oof, ptest, _ = oof_sensor_predictions(
                train_df,
                test_df,
                sensor,
                factory,
                global_classes,
                global_class_to_id,
                random_state=RANDOM_STATE + repeat,
            )

            train_parts.append(oof)
            test_parts.append(ptest)

    for sensor in combo:
        train_parts.append(train_df[[f"has_{sensor}"]].values.astype(float))
        test_parts.append(test_df[[f"has_{sensor}"]].values.astype(float))

    X_train_meta = np.hstack(train_parts)
    X_test_meta = np.hstack(test_parts)

    y_global = class_names_to_global_ids(
        train_df["class_name"].values,
        global_class_to_id,
    )

    y_local, local_classes = local_encode(y_global)

    meta = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            random_state=RANDOM_STATE + repeat,
        )),
    ])

    meta.fit(X_train_meta, y_local)

    p_global = predict_proba_global_from_local_model(
        meta,
        X_test_meta,
        local_classes,
        n_global,
    )

    y_pred = np.array([global_classes[i] for i in np.argmax(p_global, axis=1)])
    y_true = test_df["class_name"].values

    m = metrics_multiclass(y_true, y_pred, global_classes)
    m.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_stacking",
        "model": "logreg_meta",
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(X_train_meta.shape[1]),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    preds = test_df[["sample_key", "class_name"]].copy()
    preds["repeat"] = repeat
    preds["combo"] = combo_name
    preds["mode"] = "late_stacking"
    preds["model"] = "logreg_meta"
    preds["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        preds[f"prob_{cls}"] = p_global[:, i]

    return m, preds


# ============================================================
# SUMÁRIO / PLOTS
# ============================================================

def summarize_metrics(metrics_df):
    metric_cols = [
        "accuracy",
        "balanced_accuracy",
        "macro_precision",
        "macro_recall",
        "macro_f1",
        "weighted_f1",
    ]

    agg = {}

    for m in metric_cols:
        agg[f"{m}_mean"] = (m, "mean")
        agg[f"{m}_std"] = (m, "std")

    agg["n_samples_mean"] = ("n_samples", "mean")
    agg["n_test_mean"] = ("n_test", "mean")
    agg["n_features_mean"] = ("n_features", "mean")

    out = (
        metrics_df
        .groupby(["combo", "mode", "model"])
        .agg(**agg)
        .reset_index()
        .sort_values(
            ["balanced_accuracy_mean", "macro_f1_mean", "weighted_f1_mean"],
            ascending=False,
        )
    )

    return out


def save_confusion_matrix(y_true, y_pred, labels, out_path, title):
    cm = confusion_matrix(y_true, y_pred, labels=labels)

    plt.figure(figsize=(8, 7))
    plt.imshow(cm)
    plt.title(title)
    plt.xticks(range(len(labels)), labels, rotation=45, ha="right")
    plt.yticks(range(len(labels)), labels)
    plt.xlabel("Predito")
    plt.ylabel("Real")

    for i in range(len(labels)):
        for j in range(len(labels)):
            plt.text(j, i, str(cm[i, j]), ha="center", va="center")

    plt.tight_layout()
    plt.savefig(out_path, dpi=200)
    plt.close()


# ============================================================
# MAIN
# ============================================================

def main():

    print("=" * 80)
    print("CLASSIFICAÇÃO MULTICLASSE CORRIGIDA COM IR + V + AUDIO")
    print("=" * 80)

    tables = load_sensor_tables()

    class_availability = print_class_availability(tables)

    global_classes = sorted(class_availability["class_name"].unique().tolist())
    global_class_to_id = {c: i for i, c in enumerate(global_classes)}

    print("\nClasses globais:")
    for i, cls in enumerate(global_classes):
        print(f"  {i}: {cls}")

    with open(OUT_DIR / "global_classes.json", "w", encoding="utf-8") as f:
        json.dump(global_classes, f, indent=2)

    all_metrics = []
    all_preds = []
    combo_sizes = []

    for combo in COMBOS:

        combo_name = "+".join(combo)

        print("\n" + "#" * 80)
        print(f"COMBINAÇÃO: {combo_name}")
        print("#" * 80)

        df_combo = build_union_dataset(tables, combo)

        counts = df_combo["class_name"].value_counts().sort_index()

        print("Amostras:", len(df_combo))
        print("Classes:")
        print(counts.to_string())

        combo_sizes.append({
            "combo": combo_name,
            "n_samples": int(len(df_combo)),
            "classes": "|".join(counts.index.tolist()),
            **{f"n_{cls}": int(counts.get(cls, 0)) for cls in global_classes},
        })

        if df_combo["class_name"].nunique() < 2:
            print("[SKIP] Apenas uma classe.")
            continue

        for repeat in range(N_REPEATS):

            seed = RANDOM_STATE + repeat

            print(f"\n{combo_name} | repetição {repeat + 1}/{N_REPEATS}")

            try:
                train_df, test_df = safe_stratified_split(df_combo, random_state=seed)
            except Exception as e:
                print("[SKIP]", e)
                continue

            models = model_zoo(random_state=seed)

            # ------------------------------------------------
            # 1) Unimodal
            # ------------------------------------------------
            for sensor in combo:

                x_cols = get_feature_cols(train_df, sensor)

                tr_s = train_df[train_df[f"has_{sensor}"] == 1].copy()
                te_s = test_df[test_df[f"has_{sensor}"] == 1].copy()

                if len(tr_s) < 5 or len(te_s) < 2:
                    continue

                if tr_s["class_name"].nunique() < 2:
                    continue

                for model_name, factory in models.items():

                    try:
                        m, p = evaluate_direct(
                            tr_s,
                            te_s,
                            x_cols,
                            model_name,
                            factory,
                            combo_name,
                            f"unimodal_{sensor}",
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                    except Exception as e:
                        print(f"[ERRO] {combo_name} unimodal {sensor} {model_name}: {e}")

            # ------------------------------------------------
            # 2) Early fusion com sensores ausentes
            # ------------------------------------------------
            if len(combo) >= 2:

                x_cols = get_combo_feature_cols(train_df, combo)

                for model_name, factory in models.items():

                    try:
                        m, p = evaluate_direct(
                            train_df,
                            test_df,
                            x_cols,
                            model_name,
                            factory,
                            combo_name,
                            "early_concat_missing",
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                    except Exception as e:
                        print(f"[ERRO] {combo_name} early {model_name}: {e}")

            # ------------------------------------------------
            # 3) Late fusion por modelo
            # ------------------------------------------------
            if len(combo) >= 2:

                for model_name, factory in models.items():

                    try:
                        metrics, preds = evaluate_late_fusion(
                            train_df,
                            test_df,
                            combo,
                            combo_name,
                            model_name,
                            factory,
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.extend(metrics)
                        all_preds.extend(preds)

                    except Exception as e:
                        print(f"[ERRO] {combo_name} late {model_name}: {e}")

                # ------------------------------------------------
                # 4) Late stacking
                # ------------------------------------------------
                try:
                    stacking_models = {
                        k: v for k, v in models.items()
                        if k in [
                            "logreg_l2",
                            "linear_svm",
                            "random_forest",
                            "extra_trees",
                            "hist_gb",
                            "lightgbm",
                            "catboost",
                        ]
                    }

                    m, p = evaluate_late_stacking(
                        train_df,
                        test_df,
                        combo,
                        combo_name,
                        stacking_models,
                        repeat,
                        global_classes,
                        global_class_to_id,
                    )

                    all_metrics.append(m)
                    all_preds.append(p)

                except Exception as e:
                    print(f"[ERRO] {combo_name} stacking: {e}")

    # ========================================================
    # SALVAR
    # ========================================================

    combo_sizes_df = pd.DataFrame(combo_sizes)
    combo_sizes_path = OUT_DIR / "combination_sizes_multiclass.csv"
    combo_sizes_df.to_csv(combo_sizes_path, index=False, encoding="utf-8-sig")

    metrics_df = pd.DataFrame(all_metrics)
    metrics_path = OUT_DIR / "metrics_all_repeats_multiclass.csv"
    metrics_df.to_csv(metrics_path, index=False, encoding="utf-8-sig")

    summary_df = summarize_metrics(metrics_df) if not metrics_df.empty else pd.DataFrame()
    summary_path = OUT_DIR / "metrics_summary_multiclass.csv"
    summary_df.to_csv(summary_path, index=False, encoding="utf-8-sig")

    preds_df = pd.concat(all_preds, ignore_index=True) if all_preds else pd.DataFrame()
    preds_path = OUT_DIR / "predictions_all_multiclass.csv"
    preds_df.to_csv(preds_path, index=False, encoding="utf-8-sig")

    if not summary_df.empty and not preds_df.empty:

        best = summary_df.iloc[0]

        best_preds = preds_df[
            (preds_df["combo"] == best["combo"]) &
            (preds_df["mode"] == best["mode"]) &
            (preds_df["model"] == best["model"])
        ].copy()

        cm_path = OUT_DIR / "confusion_matrix_best_model.png"

        save_confusion_matrix(
            best_preds["class_name"].values,
            best_preds["pred_class"].values,
            global_classes,
            cm_path,
            f"{best['combo']} | {best['mode']} | {best['model']}"
        )

    # ========================================================
    # PRINT FINAL
    # ========================================================

    print("\n" + "=" * 80)
    print("TAMANHO DAS COMBINAÇÕES")
    print("=" * 80)
    print(combo_sizes_df.to_string(index=False))

    print("\n" + "=" * 80)
    print("TOP RESULTADOS MULTICLASSE")
    print("=" * 80)

    if not summary_df.empty:

        show_cols = [
            "combo",
            "mode",
            "model",
            "balanced_accuracy_mean",
            "balanced_accuracy_std",
            "macro_f1_mean",
            "macro_f1_std",
            "weighted_f1_mean",
            "weighted_f1_std",
            "n_samples_mean",
            "n_features_mean",
        ]

        show_cols = [c for c in show_cols if c in summary_df.columns]

        print(summary_df[show_cols].head(40).to_string(index=False))

    else:
        print("[AVISO] Nenhum resultado gerado.")

    print("\n" + "=" * 80)
    print("ARQUIVOS SALVOS")
    print("=" * 80)
    print(f"[OK] {combo_sizes_path}")
    print(f"[OK] {metrics_path}")
    print(f"[OK] {summary_path}")
    print(f"[OK] {preds_path}")
    print(f"[OK] {OUT_DIR / 'global_classes.json'}")
    print(f"[OK] Diretório:\n{OUT_DIR}")


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# multiclass_fusion_with_audio_FIXED.py
# ============================================================
# Classificação multiclasse com fusão de sensores:
#
# Sensores:
#   IR      -> vídeo infravermelho
#   V       -> vídeo visível
#   AUDIO   -> áudio
#
# Classes globais:
#   AIRPLANE
#   BACKGROUND
#   BIRD
#   DRONE
#   HELICOPTER
#
# Avaliações:
#   - unimodal
#   - early fusion por concatenação
#   - late fusion por média
#   - late fusion ponderada por OOF
#   - late fusion por stacking
#
# Saídas:
#   C:\Users\Luis\Downloads\exploracao_drone_dataset\multiclass_fusion_audio_results_fixed
# ============================================================

from pathlib import Path
import sys
import subprocess
import warnings
import re
import json
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ============================================================
# INSTALAÇÃO / IMPORTS
# ============================================================

def ensure_package(import_name, pip_name=None, optional=False):
    try:
        return __import__(import_name)
    except Exception:
        if optional:
            return None
        pkg = pip_name if pip_name is not None else import_name
        print(f"[INFO] Instalando pacote ausente: {pkg}")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])
        return __import__(import_name)


ensure_package("numpy")
ensure_package("pandas")
ensure_package("sklearn", "scikit-learn")
ensure_package("matplotlib")

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
)
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix,
)
import matplotlib.pyplot as plt


# ============================================================
# CONFIGURAÇÕES
# ============================================================

FEATURE_ROOT = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\fusion_3sensors_baseline"
)

FEATURE_FILES = {
    "IR": FEATURE_ROOT / "features_IR.csv",
    "V": FEATURE_ROOT / "features_V.csv",
    "AUDIO": FEATURE_ROOT / "features_AUDIO.csv",
}

OUT_DIR = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\multiclass_fusion_audio_results_fixed"
)
OUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42

# Para teste rápido, use 3.
# Para resultado final, use 10 ou 20.
N_REPEATS = 10

TEST_SIZE = 0.30
MAX_CV_FOLDS = 5

# Se quiser instalar automaticamente XGBoost/LightGBM/CatBoost, coloque True.
INSTALL_OPTIONAL_SOTA = False

SENSOR_LIST = ["IR", "V", "AUDIO"]

COMBOS = [
    ("IR",),
    ("V",),
    ("AUDIO",),
    ("IR", "V"),
    ("IR", "AUDIO"),
    ("V", "AUDIO"),
    ("IR", "V", "AUDIO"),
]


# ============================================================
# FEATURES / CLASSES
# ============================================================

def normalize_class_name(x):
    return str(x).upper().strip()


def get_feature_cols(df, sensor=None):
    """
    Retorna colunas de features.
    Aceita:
      IR_f0000
      V_f0000
      audio_f0000
      AUDIO_f0000
    """
    if sensor is None:
        return [
            c for c in df.columns
            if re.search(r"_f\d+$", c, flags=re.IGNORECASE)
        ]

    prefix = f"{sensor}_f".upper()
    return [
        c for c in df.columns
        if c.upper().startswith(prefix)
    ]


def standardize_feature_columns(df, sensor):
    """
    Padroniza nomes de features:
      audio_f0000 -> AUDIO_f0000
      IR_f0000    -> IR_f0000
      V_f0000     -> V_f0000
    """
    rename = {}

    for c in df.columns:
        if re.search(r"_f\d+$", c, flags=re.IGNORECASE):
            m = re.search(r"_f(\d+)$", c, flags=re.IGNORECASE)
            if m:
                num = int(m.group(1))
                # aceita apenas colunas do sensor
                if c.upper().startswith(f"{sensor}_F") or (
                    sensor == "AUDIO" and c.upper().startswith("AUDIO_F")
                ):
                    rename[c] = f"{sensor}_f{num:04d}"

    if rename:
        df = df.rename(columns=rename)

    return df


def force_numeric_features(df, feature_cols):
    """
    Converte features para float, substituindo texto inválido por NaN.
    """
    df = df.copy()

    for c in feature_cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df


def load_sensor_tables():
    tables = {}

    for sensor, path in FEATURE_FILES.items():

        if not path.exists():
            raise FileNotFoundError(f"Arquivo não encontrado: {path}")

        df = pd.read_csv(path)

        required = {"sample_key", "class_name"}
        missing = required - set(df.columns)

        if missing:
            raise ValueError(f"{path} sem colunas obrigatórias: {missing}")

        df["sample_key"] = df["sample_key"].astype(str)
        df["class_name"] = df["class_name"].apply(normalize_class_name)

        # Corrige audio_f0000 -> AUDIO_f0000
        df = standardize_feature_columns(df, sensor)

        fcols = get_feature_cols(df, sensor)

        if len(fcols) == 0:
            raise ValueError(f"Nenhuma feature encontrada para sensor {sensor}")

        df = force_numeric_features(df, fcols)

        # Mantém só sample_key, class_name e features
        df = df[["sample_key", "class_name"] + fcols].drop_duplicates("sample_key")

        tables[sensor] = df

        print(
            f"[OK] {sensor}: {df.shape[0]} amostras, "
            f"{len(fcols)} features, classes={sorted(df['class_name'].unique())}"
        )

    return tables


def print_class_availability(tables):
    rows = []

    for sensor, df in tables.items():
        for cls, g in df.groupby("class_name"):
            rows.append({
                "sensor": sensor,
                "class_name": cls,
                "n_samples": int(g["sample_key"].nunique()),
            })

    out = pd.DataFrame(rows).sort_values(["sensor", "class_name"])
    path = OUT_DIR / "class_availability_by_sensor.csv"
    out.to_csv(path, index=False, encoding="utf-8-sig")

    print("\n" + "=" * 80)
    print("CLASSES DISPONÍVEIS POR SENSOR")
    print("=" * 80)
    print(out.to_string(index=False))
    print(f"[OK] Salvo em: {path}")

    return out


def build_union_dataset(tables, combo):
    """
    União de amostras:
    entra qualquer sample_key que exista em pelo menos um sensor da combinação.

    Features de sensores ausentes ficam NaN.
    Adiciona has_IR, has_V, has_AUDIO.
    """

    meta_rows = []

    for sensor in combo:
        df = tables[sensor]
        for _, r in df[["sample_key", "class_name"]].iterrows():
            meta_rows.append({
                "sample_key": r["sample_key"],
                "class_name": r["class_name"],
                "sensor": sensor,
            })

    meta = pd.DataFrame(meta_rows)

    # Classe por sample_key
    # Se aparecer em mais de um sensor, usa a moda.
    class_by_sample = (
        meta.groupby("sample_key")["class_name"]
        .agg(lambda x: x.value_counts().index[0])
        .reset_index()
    )

    merged = class_by_sample.copy()

    for sensor in combo:
        df = tables[sensor].copy()
        fcols = get_feature_cols(df, sensor)

        sensor_df = df[["sample_key"] + fcols].drop_duplicates("sample_key")

        merged = merged.merge(sensor_df, on="sample_key", how="left")

        merged[f"has_{sensor}"] = (
            merged[fcols].notna().any(axis=1).astype(int)
        )

    return merged


def get_combo_feature_cols(df, combo):
    cols = []

    for sensor in combo:
        cols.extend(get_feature_cols(df, sensor))

    for sensor in combo:
        has_col = f"has_{sensor}"
        if has_col in df.columns:
            cols.append(has_col)

    return cols


# ============================================================
# MODELOS
# ============================================================

def make_pipeline(estimator, scale=True):
    steps = [("imputer", SimpleImputer(strategy="median"))]

    if scale:
        steps.append(("scaler", StandardScaler()))

    steps.append(("clf", estimator))
    return Pipeline(steps)


def model_zoo(random_state=42):
    models = {}

    models["dummy_prior"] = lambda: make_pipeline(
        DummyClassifier(strategy="prior", random_state=random_state),
        scale=False,
    )

    # Corrigido: sem multi_class
    models["logreg_l2"] = lambda: make_pipeline(
        LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            solver="lbfgs",
            random_state=random_state,
        ),
        scale=True,
    )

    models["linear_svm"] = lambda: make_pipeline(
        SVC(
            kernel="linear",
            probability=True,
            class_weight="balanced",
            random_state=random_state,
        ),
        scale=True,
    )

    models["random_forest"] = lambda: make_pipeline(
        RandomForestClassifier(
            n_estimators=500,
            class_weight="balanced",
            random_state=random_state,
            n_jobs=-1,
        ),
        scale=False,
    )

    models["extra_trees"] = lambda: make_pipeline(
        ExtraTreesClassifier(
            n_estimators=700,
            class_weight="balanced",
            random_state=random_state,
            n_jobs=-1,
        ),
        scale=False,
    )

    models["hist_gb"] = lambda: make_pipeline(
        HistGradientBoostingClassifier(
            learning_rate=0.05,
            max_iter=300,
            l2_regularization=0.01,
            random_state=random_state,
        ),
        scale=False,
    )

    models["mlp"] = lambda: make_pipeline(
        MLPClassifier(
            hidden_layer_sizes=(256, 64),
            activation="relu",
            alpha=1e-4,
            learning_rate_init=1e-3,
            max_iter=800,
            early_stopping=True,
            random_state=random_state,
        ),
        scale=True,
    )

    if INSTALL_OPTIONAL_SOTA:
        for pkg in ["xgboost", "lightgbm", "catboost"]:
            try:
                __import__(pkg)
            except Exception:
                print(f"[INFO] Instalando pacote opcional: {pkg}")
                subprocess.check_call([
                    sys.executable, "-m", "pip", "install", "-q", pkg
                ])

    try:
        from xgboost import XGBClassifier

        models["xgboost"] = lambda: make_pipeline(
            XGBClassifier(
                n_estimators=500,
                max_depth=3,
                learning_rate=0.03,
                subsample=0.90,
                colsample_bytree=0.90,
                objective="multi:softprob",
                eval_metric="mlogloss",
                random_state=random_state,
                n_jobs=-1,
            ),
            scale=False,
        )
    except Exception:
        print("[INFO] XGBoost não disponível.")

    try:
        from lightgbm import LGBMClassifier

        models["lightgbm"] = lambda: make_pipeline(
            LGBMClassifier(
                n_estimators=700,
                learning_rate=0.03,
                num_leaves=31,
                subsample=0.90,
                colsample_bytree=0.90,
                class_weight="balanced",
                random_state=random_state,
                n_jobs=-1,
                verbose=-1,
            ),
            scale=False,
        )
    except Exception:
        print("[INFO] LightGBM não disponível.")

    try:
        from catboost import CatBoostClassifier

        models["catboost"] = lambda: make_pipeline(
            CatBoostClassifier(
                iterations=700,
                depth=4,
                learning_rate=0.03,
                loss_function="MultiClass",
                random_seed=random_state,
                verbose=False,
                auto_class_weights="Balanced",
            ),
            scale=False,
        )
    except Exception:
        print("[INFO] CatBoost não disponível.")

    return models


# ============================================================
# TREINO COM RÓTULOS LOCAIS
# ============================================================

def class_names_to_global_ids(y_names, global_class_to_id):
    return np.array([global_class_to_id[x] for x in y_names], dtype=int)


def local_encode(y_global):
    """
    Converte ids globais para ids locais 0..K-1.
    Necessário para XGBoost e outros modelos.
    """
    local_classes = sorted(np.unique(y_global).tolist())
    global_to_local = {g: i for i, g in enumerate(local_classes)}

    y_local = np.array([global_to_local[g] for g in y_global], dtype=int)

    return y_local, local_classes


def predict_proba_global_from_local_model(model, X, local_classes, n_global_classes):
    """
    O modelo foi treinado com labels locais.
    Aqui convertemos as probabilidades para o espaço global.
    """
    if hasattr(model, "predict_proba"):
        p_local = model.predict_proba(X)
    else:
        y_pred_local = model.predict(X)
        p_local = np.zeros((len(y_pred_local), len(local_classes)), dtype=float)
        for i, yy in enumerate(y_pred_local):
            p_local[i, int(yy)] = 1.0

    out = np.zeros((p_local.shape[0], n_global_classes), dtype=float)

    model_classes = model.named_steps["clf"].classes_

    for j, local_label in enumerate(model_classes):
        local_label = int(local_label)
        global_id = local_classes[local_label]
        out[:, global_id] = p_local[:, j]

    row_sum = out.sum(axis=1, keepdims=True)
    row_sum[row_sum <= 1e-12] = 1.0
    out = out / row_sum

    return out


def fit_model_local_labels(train_df, x_cols, model_factory, global_class_to_id):
    X_train = train_df[x_cols].apply(pd.to_numeric, errors="coerce").values.astype(float)

    y_global = class_names_to_global_ids(
        train_df["class_name"].values,
        global_class_to_id,
    )

    y_local, local_classes = local_encode(y_global)

    model = model_factory()
    model.fit(X_train, y_local)

    return model, local_classes


def predict_model_global(model, local_classes, test_df, x_cols, n_global_classes):
    X_test = test_df[x_cols].apply(pd.to_numeric, errors="coerce").values.astype(float)

    return predict_proba_global_from_local_model(
        model,
        X_test,
        local_classes,
        n_global_classes,
    )


# ============================================================
# MÉTRICAS
# ============================================================

def metrics_multiclass(y_true_names, y_pred_names, labels):
    return {
        "accuracy": float(accuracy_score(y_true_names, y_pred_names)),
        "balanced_accuracy": float(balanced_accuracy_score(y_true_names, y_pred_names)),
        "macro_precision": float(
            precision_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_recall": float(
            recall_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="macro",
                zero_division=0,
            )
        ),
        "macro_f1": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="macro",
                zero_division=0,
            )
        ),
        "weighted_f1": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=labels,
                average="weighted",
                zero_division=0,
            )
        ),
    }


def safe_stratified_split(df, random_state):
    counts = df["class_name"].value_counts()

    # Remove classes com menos de 2 amostras
    valid = counts[counts >= 2].index.tolist()
    df = df[df["class_name"].isin(valid)].copy()

    if df["class_name"].nunique() < 2:
        raise ValueError("Menos de duas classes com amostras suficientes.")

    train_df, test_df = train_test_split(
        df,
        test_size=TEST_SIZE,
        random_state=random_state,
        stratify=df["class_name"],
        shuffle=True,
    )

    return train_df.reset_index(drop=True), test_df.reset_index(drop=True)


# ============================================================
# AVALIAÇÃO DIRETA
# ============================================================

def evaluate_direct(
    train_df,
    test_df,
    x_cols,
    model_name,
    model_factory,
    combo_name,
    mode,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    model, local_classes = fit_model_local_labels(
        train_df,
        x_cols,
        model_factory,
        global_class_to_id,
    )

    p_global = predict_model_global(
        model,
        local_classes,
        test_df,
        x_cols,
        n_global,
    )

    pred_ids = np.argmax(p_global, axis=1)
    y_pred = np.array([global_classes[i] for i in pred_ids])
    y_true = test_df["class_name"].values

    m = metrics_multiclass(y_true, y_pred, global_classes)

    m.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": mode,
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(x_cols)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    preds = test_df[["sample_key", "class_name"]].copy()
    preds["repeat"] = repeat
    preds["combo"] = combo_name
    preds["mode"] = mode
    preds["model"] = model_name
    preds["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        preds[f"prob_{cls}"] = p_global[:, i]

    return m, preds


# ============================================================
# LATE FUSION
# ============================================================

def oof_sensor_predictions(
    train_df,
    test_df,
    sensor,
    model_factory,
    global_classes,
    global_class_to_id,
    random_state,
):
    fcols = get_feature_cols(train_df, sensor)
    has_col = f"has_{sensor}"

    n_global = len(global_classes)

    oof = np.zeros((len(train_df), n_global), dtype=float)
    test_prob = np.zeros((len(test_df), n_global), dtype=float)

    train_idx = np.where(train_df[has_col].values == 1)[0]
    test_idx = np.where(test_df[has_col].values == 1)[0]

    if len(train_idx) < 4:
        return oof, test_prob, 0.0

    tr_sensor = train_df.iloc[train_idx].copy()

    if tr_sensor["class_name"].nunique() < 2:
        return oof, test_prob, 0.0

    y_names = tr_sensor["class_name"].values
    y_global = class_names_to_global_ids(y_names, global_class_to_id)

    min_count = pd.Series(y_names).value_counts().min()
    n_splits = min(MAX_CV_FOLDS, int(min_count))

    if n_splits < 2:
        return oof, test_prob, 0.0

    X_sensor = tr_sensor[fcols].apply(pd.to_numeric, errors="coerce").values.astype(float)

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state,
    )

    for tr, va in skf.split(X_sensor, y_global):
        fold_df_train = tr_sensor.iloc[tr].copy()
        fold_df_val = tr_sensor.iloc[va].copy()

        model, local_classes = fit_model_local_labels(
            fold_df_train,
            fcols,
            model_factory,
            global_class_to_id,
        )

        p_val = predict_model_global(
            model,
            local_classes,
            fold_df_val,
            fcols,
            n_global,
        )

        original_val_idx = train_idx[va]
        oof[original_val_idx, :] = p_val

    # Modelo final
    model, local_classes = fit_model_local_labels(
        tr_sensor,
        fcols,
        model_factory,
        global_class_to_id,
    )

    if len(test_idx) > 0:
        te_sensor = test_df.iloc[test_idx].copy()

        p_test = predict_model_global(
            model,
            local_classes,
            te_sensor,
            fcols,
            n_global,
        )

        test_prob[test_idx, :] = p_test

    # Peso OOF por balanced accuracy
    available_oof = oof[train_idx]
    pred_ids = np.argmax(available_oof, axis=1)
    y_pred = np.array([global_classes[i] for i in pred_ids])
    y_true = tr_sensor["class_name"].values

    try:
        weight = balanced_accuracy_score(y_true, y_pred)
    except Exception:
        weight = 0.0

    return oof, test_prob, float(weight)


def normalize_rows(p):
    p = np.asarray(p, dtype=float)
    s = p.sum(axis=1, keepdims=True)
    s[s <= 1e-12] = 1.0
    return p / s


def evaluate_late_fusion(
    train_df,
    test_df,
    combo,
    combo_name,
    model_name,
    model_factory,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    sensor_test = {}
    sensor_weights = {}

    for sensor in combo:
        _, ptest, weight = oof_sensor_predictions(
            train_df,
            test_df,
            sensor,
            model_factory,
            global_classes,
            global_class_to_id,
            random_state=RANDOM_STATE + repeat,
        )

        sensor_test[sensor] = ptest
        sensor_weights[sensor] = weight

    y_true = test_df["class_name"].values

    # ----------------------------
    # Média simples
    # ----------------------------
    p_sum = np.zeros((len(test_df), n_global), dtype=float)
    count = np.zeros((len(test_df), 1), dtype=float)

    for sensor in combo:
        has = test_df[f"has_{sensor}"].values.reshape(-1, 1).astype(float)
        p_sum += sensor_test[sensor] * has
        count += has

    count[count <= 1e-12] = 1.0
    p_mean = normalize_rows(p_sum / count)

    y_pred = np.array([global_classes[i] for i in np.argmax(p_mean, axis=1)])

    m_mean = metrics_multiclass(y_true, y_pred, global_classes)
    m_mean.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_mean",
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(combo)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    pred_mean = test_df[["sample_key", "class_name"]].copy()
    pred_mean["repeat"] = repeat
    pred_mean["combo"] = combo_name
    pred_mean["mode"] = "late_mean"
    pred_mean["model"] = model_name
    pred_mean["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        pred_mean[f"prob_{cls}"] = p_mean[:, i]

    # ----------------------------
    # Média ponderada
    # ----------------------------
    p_wsum = np.zeros((len(test_df), n_global), dtype=float)
    w_sum = np.zeros((len(test_df), 1), dtype=float)

    for sensor in combo:
        has = test_df[f"has_{sensor}"].values.reshape(-1, 1).astype(float)
        w = max(sensor_weights.get(sensor, 0.0), 1e-6)

        p_wsum += sensor_test[sensor] * has * w
        w_sum += has * w

    w_sum[w_sum <= 1e-12] = 1.0
    p_weighted = normalize_rows(p_wsum / w_sum)

    y_pred_w = np.array([global_classes[i] for i in np.argmax(p_weighted, axis=1)])

    m_weighted = metrics_multiclass(y_true, y_pred_w, global_classes)
    m_weighted.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_weighted_oof",
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(combo)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    pred_weighted = test_df[["sample_key", "class_name"]].copy()
    pred_weighted["repeat"] = repeat
    pred_weighted["combo"] = combo_name
    pred_weighted["mode"] = "late_weighted_oof"
    pred_weighted["model"] = model_name
    pred_weighted["pred_class"] = y_pred_w

    for i, cls in enumerate(global_classes):
        pred_weighted[f"prob_{cls}"] = p_weighted[:, i]

    return [m_mean, m_weighted], [pred_mean, pred_weighted]


def evaluate_late_stacking(
    train_df,
    test_df,
    combo,
    combo_name,
    base_models,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    train_parts = []
    test_parts = []

    for model_name, factory in base_models.items():
        for sensor in combo:
            oof, ptest, _ = oof_sensor_predictions(
                train_df,
                test_df,
                sensor,
                factory,
                global_classes,
                global_class_to_id,
                random_state=RANDOM_STATE + repeat,
            )

            train_parts.append(oof)
            test_parts.append(ptest)

    for sensor in combo:
        train_parts.append(train_df[[f"has_{sensor}"]].values.astype(float))
        test_parts.append(test_df[[f"has_{sensor}"]].values.astype(float))

    X_train_meta = np.hstack(train_parts)
    X_test_meta = np.hstack(test_parts)

    y_global = class_names_to_global_ids(
        train_df["class_name"].values,
        global_class_to_id,
    )

    y_local, local_classes = local_encode(y_global)

    meta = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            random_state=RANDOM_STATE + repeat,
        )),
    ])

    meta.fit(X_train_meta, y_local)

    p_global = predict_proba_global_from_local_model(
        meta,
        X_test_meta,
        local_classes,
        n_global,
    )

    y_pred = np.array([global_classes[i] for i in np.argmax(p_global, axis=1)])
    y_true = test_df["class_name"].values

    m = metrics_multiclass(y_true, y_pred, global_classes)
    m.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_stacking",
        "model": "logreg_meta",
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(X_train_meta.shape[1]),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    preds = test_df[["sample_key", "class_name"]].copy()
    preds["repeat"] = repeat
    preds["combo"] = combo_name
    preds["mode"] = "late_stacking"
    preds["model"] = "logreg_meta"
    preds["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        preds[f"prob_{cls}"] = p_global[:, i]

    return m, preds


# ============================================================
# SUMÁRIO / PLOTS
# ============================================================

def summarize_metrics(metrics_df):
    metric_cols = [
        "accuracy",
        "balanced_accuracy",
        "macro_precision",
        "macro_recall",
        "macro_f1",
        "weighted_f1",
    ]

    agg = {}

    for m in metric_cols:
        agg[f"{m}_mean"] = (m, "mean")
        agg[f"{m}_std"] = (m, "std")

    agg["n_samples_mean"] = ("n_samples", "mean")
    agg["n_test_mean"] = ("n_test", "mean")
    agg["n_features_mean"] = ("n_features", "mean")

    out = (
        metrics_df
        .groupby(["combo", "mode", "model"])
        .agg(**agg)
        .reset_index()
        .sort_values(
            ["balanced_accuracy_mean", "macro_f1_mean", "weighted_f1_mean"],
            ascending=False,
        )
    )

    return out


def save_confusion_matrix(y_true, y_pred, labels, out_path, title):
    cm = confusion_matrix(y_true, y_pred, labels=labels)

    plt.figure(figsize=(8, 7))
    plt.imshow(cm)
    plt.title(title)
    plt.xticks(range(len(labels)), labels, rotation=45, ha="right")
    plt.yticks(range(len(labels)), labels)
    plt.xlabel("Predito")
    plt.ylabel("Real")

    for i in range(len(labels)):
        for j in range(len(labels)):
            plt.text(j, i, str(cm[i, j]), ha="center", va="center")

    plt.tight_layout()
    plt.savefig(out_path, dpi=200)
    plt.close()


# ============================================================
# MAIN
# ============================================================

def main():

    print("=" * 80)
    print("CLASSIFICAÇÃO MULTICLASSE CORRIGIDA COM IR + V + AUDIO")
    print("=" * 80)

    tables = load_sensor_tables()

    class_availability = print_class_availability(tables)

    global_classes = sorted(class_availability["class_name"].unique().tolist())
    global_class_to_id = {c: i for i, c in enumerate(global_classes)}

    print("\nClasses globais:")
    for i, cls in enumerate(global_classes):
        print(f"  {i}: {cls}")

    with open(OUT_DIR / "global_classes.json", "w", encoding="utf-8") as f:
        json.dump(global_classes, f, indent=2)

    all_metrics = []
    all_preds = []
    combo_sizes = []

    for combo in COMBOS:

        combo_name = "+".join(combo)

        print("\n" + "#" * 80)
        print(f"COMBINAÇÃO: {combo_name}")
        print("#" * 80)

        df_combo = build_union_dataset(tables, combo)

        counts = df_combo["class_name"].value_counts().sort_index()

        print("Amostras:", len(df_combo))
        print("Classes:")
        print(counts.to_string())

        combo_sizes.append({
            "combo": combo_name,
            "n_samples": int(len(df_combo)),
            "classes": "|".join(counts.index.tolist()),
            **{f"n_{cls}": int(counts.get(cls, 0)) for cls in global_classes},
        })

        if df_combo["class_name"].nunique() < 2:
            print("[SKIP] Apenas uma classe.")
            continue

        for repeat in range(N_REPEATS):

            seed = RANDOM_STATE + repeat

            print(f"\n{combo_name} | repetição {repeat + 1}/{N_REPEATS}")

            try:
                train_df, test_df = safe_stratified_split(df_combo, random_state=seed)
            except Exception as e:
                print("[SKIP]", e)
                continue

            models = model_zoo(random_state=seed)

            # ------------------------------------------------
            # 1) Unimodal
            # ------------------------------------------------
            for sensor in combo:

                x_cols = get_feature_cols(train_df, sensor)

                tr_s = train_df[train_df[f"has_{sensor}"] == 1].copy()
                te_s = test_df[test_df[f"has_{sensor}"] == 1].copy()

                if len(tr_s) < 5 or len(te_s) < 2:
                    continue

                if tr_s["class_name"].nunique() < 2:
                    continue

                for model_name, factory in models.items():

                    try:
                        m, p = evaluate_direct(
                            tr_s,
                            te_s,
                            x_cols,
                            model_name,
                            factory,
                            combo_name,
                            f"unimodal_{sensor}",
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                    except Exception as e:
                        print(f"[ERRO] {combo_name} unimodal {sensor} {model_name}: {e}")

            # ------------------------------------------------
            # 2) Early fusion com sensores ausentes
            # ------------------------------------------------
            if len(combo) >= 2:

                x_cols = get_combo_feature_cols(train_df, combo)

                for model_name, factory in models.items():

                    try:
                        m, p = evaluate_direct(
                            train_df,
                            test_df,
                            x_cols,
                            model_name,
                            factory,
                            combo_name,
                            "early_concat_missing",
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                    except Exception as e:
                        print(f"[ERRO] {combo_name} early {model_name}: {e}")

            # ------------------------------------------------
            # 3) Late fusion por modelo
            # ------------------------------------------------
            if len(combo) >= 2:

                for model_name, factory in models.items():

                    try:
                        metrics, preds = evaluate_late_fusion(
                            train_df,
                            test_df,
                            combo,
                            combo_name,
                            model_name,
                            factory,
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.extend(metrics)
                        all_preds.extend(preds)

                    except Exception as e:
                        print(f"[ERRO] {combo_name} late {model_name}: {e}")

                # ------------------------------------------------
                # 4) Late stacking
                # ------------------------------------------------
                try:
                    stacking_models = {
                        k: v for k, v in models.items()
                        if k in [
                            "logreg_l2",
                            "linear_svm",
                            "random_forest",
                            "extra_trees",
                            "hist_gb",
                            "lightgbm",
                            "catboost",
                        ]
                    }

                    m, p = evaluate_late_stacking(
                        train_df,
                        test_df,
                        combo,
                        combo_name,
                        stacking_models,
                        repeat,
                        global_classes,
                        global_class_to_id,
                    )

                    all_metrics.append(m)
                    all_preds.append(p)

                except Exception as e:
                    print(f"[ERRO] {combo_name} stacking: {e}")

    # ========================================================
    # SALVAR
    # ========================================================

    combo_sizes_df = pd.DataFrame(combo_sizes)
    combo_sizes_path = OUT_DIR / "combination_sizes_multiclass.csv"
    combo_sizes_df.to_csv(combo_sizes_path, index=False, encoding="utf-8-sig")

    metrics_df = pd.DataFrame(all_metrics)
    metrics_path = OUT_DIR / "metrics_all_repeats_multiclass.csv"
    metrics_df.to_csv(metrics_path, index=False, encoding="utf-8-sig")

    summary_df = summarize_metrics(metrics_df) if not metrics_df.empty else pd.DataFrame()
    summary_path = OUT_DIR / "metrics_summary_multiclass.csv"
    summary_df.to_csv(summary_path, index=False, encoding="utf-8-sig")

    preds_df = pd.concat(all_preds, ignore_index=True) if all_preds else pd.DataFrame()
    preds_path = OUT_DIR / "predictions_all_multiclass.csv"
    preds_df.to_csv(preds_path, index=False, encoding="utf-8-sig")

    if not summary_df.empty and not preds_df.empty:

        best = summary_df.iloc[0]

        best_preds = preds_df[
            (preds_df["combo"] == best["combo"]) &
            (preds_df["mode"] == best["mode"]) &
            (preds_df["model"] == best["model"])
        ].copy()

        cm_path = OUT_DIR / "confusion_matrix_best_model.png"

        save_confusion_matrix(
            best_preds["class_name"].values,
            best_preds["pred_class"].values,
            global_classes,
            cm_path,
            f"{best['combo']} | {best['mode']} | {best['model']}"
        )

    # ========================================================
    # PRINT FINAL
    # ========================================================

    print("\n" + "=" * 80)
    print("TAMANHO DAS COMBINAÇÕES")
    print("=" * 80)
    print(combo_sizes_df.to_string(index=False))

    print("\n" + "=" * 80)
    print("TOP RESULTADOS MULTICLASSE")
    print("=" * 80)

    if not summary_df.empty:

        show_cols = [
            "combo",
            "mode",
            "model",
            "balanced_accuracy_mean",
            "balanced_accuracy_std",
            "macro_f1_mean",
            "macro_f1_std",
            "weighted_f1_mean",
            "weighted_f1_std",
            "n_samples_mean",
            "n_features_mean",
        ]

        show_cols = [c for c in show_cols if c in summary_df.columns]

        print(summary_df[show_cols].head(40).to_string(index=False))

    else:
        print("[AVISO] Nenhum resultado gerado.")

    print("\n" + "=" * 80)
    print("ARQUIVOS SALVOS")
    print("=" * 80)
    print(f"[OK] {combo_sizes_path}")
    print(f"[OK] {metrics_path}")
    print(f"[OK] {summary_path}")
    print(f"[OK] {preds_path}")
    print(f"[OK] {OUT_DIR / 'global_classes.json'}")
    print(f"[OK] Diretório:\n{OUT_DIR}")


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# multiclass_fusion_with_audio_FIXED_METRICS.py
# ============================================================
# Classificação multiclasse com fusão de sensores:
#
# Sensores:
#   IR      -> vídeo infravermelho
#   V       -> vídeo visível
#   AUDIO   -> áudio
#
# Classes globais:
#   AIRPLANE
#   BACKGROUND
#   BIRD
#   DRONE
#   HELICOPTER
#
# Avaliações:
#   - unimodal
#   - early fusion por concatenação
#   - late fusion por média
#   - late fusion ponderada por OOF
#   - late fusion por stacking
#
# Correções principais:
#   - macro_f1 calculado apenas nas classes presentes no teste
#   - macro_f1_global salvo apenas para auditoria
#   - warnings do LightGBM silenciados
#   - modelos opcionais realmente controlados por USE_OPTIONAL_SOTA
#
# Saídas:
#   C:\Users\Luis\Downloads\exploracao_drone_dataset\multiclass_fusion_audio_results_fixed
# ============================================================

from pathlib import Path
import sys
import subprocess
import warnings
import re
import json
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

warnings.filterwarnings(
    "ignore",
    message="X does not have valid feature names.*",
    category=UserWarning,
)

warnings.filterwarnings("ignore", category=FutureWarning)


# ============================================================
# INSTALAÇÃO / IMPORTS
# ============================================================

def ensure_package(import_name, pip_name=None, optional=False):
    try:
        return __import__(import_name)
    except Exception:
        if optional:
            return None

        pkg = pip_name if pip_name is not None else import_name
        print(f"[INFO] Instalando pacote ausente: {pkg}")
        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            pkg,
        ])
        return __import__(import_name)


ensure_package("numpy")
ensure_package("pandas")
ensure_package("sklearn", "scikit-learn")
ensure_package("matplotlib")

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
)
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix,
)

import matplotlib.pyplot as plt


# ============================================================
# CONFIGURAÇÕES
# ============================================================

FEATURE_ROOT = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\fusion_3sensors_baseline"
)

FEATURE_FILES = {
    "IR": FEATURE_ROOT / "features_IR.csv",
    "V": FEATURE_ROOT / "features_V.csv",
    "AUDIO": FEATURE_ROOT / "features_AUDIO.csv",
}

OUT_DIR = Path(
    r"C:\Users\Luis\Downloads\exploracao_drone_dataset\multiclass_fusion_audio_results_fixed"
)
OUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42

# Para teste rápido, use 3 ou 5.
# Para resultado final, use 10 ou 20.
N_REPEATS = 10

TEST_SIZE = 0.30
MAX_CV_FOLDS = 5

# Controle real dos modelos opcionais.
# False = não usa XGBoost, LightGBM nem CatBoost, mesmo se estiverem instalados.
# True  = tenta usar esses modelos.
USE_OPTIONAL_SOTA = False

# Se USE_OPTIONAL_SOTA=True e quiser instalar automaticamente pacotes ausentes.
INSTALL_OPTIONAL_SOTA = False

SENSOR_LIST = ["IR", "V", "AUDIO"]

COMBOS = [
    ("IR",),
    ("V",),
    ("AUDIO",),
    ("IR", "V"),
    ("IR", "AUDIO"),
    ("V", "AUDIO"),
    ("IR", "V", "AUDIO"),
]


# ============================================================
# FEATURES / CLASSES
# ============================================================

def normalize_class_name(x):
    return str(x).upper().strip()


def get_feature_cols(df, sensor=None):
    """
    Retorna colunas de features.

    Aceita:
      IR_f0000
      V_f0000
      audio_f0000
      AUDIO_f0000
    """

    if sensor is None:
        return [
            c for c in df.columns
            if re.search(r"_f\d+$", c, flags=re.IGNORECASE)
        ]

    prefix = f"{sensor}_f".upper()

    return [
        c for c in df.columns
        if c.upper().startswith(prefix)
    ]


def standardize_feature_columns(df, sensor):
    """
    Padroniza nomes de features:
      audio_f0000 -> AUDIO_f0000
      IR_f0000    -> IR_f0000
      V_f0000     -> V_f0000
    """

    rename = {}

    for c in df.columns:
        if re.search(r"_f\d+$", c, flags=re.IGNORECASE):
            m = re.search(r"_f(\d+)$", c, flags=re.IGNORECASE)

            if m:
                num = int(m.group(1))

                if c.upper().startswith(f"{sensor}_F"):
                    rename[c] = f"{sensor}_f{num:04d}"

    if rename:
        df = df.rename(columns=rename)

    return df


def force_numeric_features(df, feature_cols):
    """
    Converte features para float, substituindo texto inválido por NaN.
    """

    df = df.copy()

    for c in feature_cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df


def load_sensor_tables():
    tables = {}

    for sensor, path in FEATURE_FILES.items():

        if not path.exists():
            raise FileNotFoundError(f"Arquivo não encontrado: {path}")

        df = pd.read_csv(path)

        required = {"sample_key", "class_name"}
        missing = required - set(df.columns)

        if missing:
            raise ValueError(f"{path} sem colunas obrigatórias: {missing}")

        df["sample_key"] = df["sample_key"].astype(str)
        df["class_name"] = df["class_name"].apply(normalize_class_name)

        df = standardize_feature_columns(df, sensor)

        fcols = get_feature_cols(df, sensor)

        if len(fcols) == 0:
            raise ValueError(f"Nenhuma feature encontrada para sensor {sensor}")

        df = force_numeric_features(df, fcols)

        df = df[["sample_key", "class_name"] + fcols].drop_duplicates("sample_key")

        tables[sensor] = df

        print(
            f"[OK] {sensor}: {df.shape[0]} amostras, "
            f"{len(fcols)} features, classes={sorted(df['class_name'].unique())}"
        )

    return tables


def print_class_availability(tables):
    rows = []

    for sensor, df in tables.items():
        for cls, g in df.groupby("class_name"):
            rows.append({
                "sensor": sensor,
                "class_name": cls,
                "n_samples": int(g["sample_key"].nunique()),
            })

    out = pd.DataFrame(rows).sort_values(["sensor", "class_name"])

    path = OUT_DIR / "class_availability_by_sensor.csv"
    out.to_csv(path, index=False, encoding="utf-8-sig")

    print("\n" + "=" * 80)
    print("CLASSES DISPONÍVEIS POR SENSOR")
    print("=" * 80)
    print(out.to_string(index=False))
    print(f"[OK] Salvo em: {path}")

    return out


def build_union_dataset(tables, combo):
    """
    União de amostras.

    Entra qualquer sample_key que exista em pelo menos um sensor da combinação.

    Features de sensores ausentes ficam NaN.

    Adiciona:
      has_IR
      has_V
      has_AUDIO
    """

    meta_rows = []

    for sensor in combo:
        df = tables[sensor]

        for _, r in df[["sample_key", "class_name"]].iterrows():
            meta_rows.append({
                "sample_key": r["sample_key"],
                "class_name": r["class_name"],
                "sensor": sensor,
            })

    meta = pd.DataFrame(meta_rows)

    class_by_sample = (
        meta.groupby("sample_key")["class_name"]
        .agg(lambda x: x.value_counts().index[0])
        .reset_index()
    )

    merged = class_by_sample.copy()

    for sensor in combo:
        df = tables[sensor].copy()

        fcols = get_feature_cols(df, sensor)

        sensor_df = df[["sample_key"] + fcols].drop_duplicates("sample_key")

        merged = merged.merge(sensor_df, on="sample_key", how="left")

        merged[f"has_{sensor}"] = (
            merged[fcols].notna().any(axis=1).astype(int)
        )

    return merged


def get_combo_feature_cols(df, combo):
    cols = []

    for sensor in combo:
        cols.extend(get_feature_cols(df, sensor))

    for sensor in combo:
        has_col = f"has_{sensor}"

        if has_col in df.columns:
            cols.append(has_col)

    return cols


# ============================================================
# MODELOS
# ============================================================

def make_pipeline(estimator, scale=True):
    steps = [
        ("imputer", SimpleImputer(strategy="median")),
    ]

    if scale:
        steps.append(("scaler", StandardScaler()))

    steps.append(("clf", estimator))

    return Pipeline(steps)


def model_zoo(random_state=42):
    models = {}

    models["dummy_prior"] = lambda: make_pipeline(
        DummyClassifier(
            strategy="prior",
            random_state=random_state,
        ),
        scale=False,
    )

    models["logreg_l2"] = lambda: make_pipeline(
        LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            solver="lbfgs",
            random_state=random_state,
        ),
        scale=True,
    )

    models["linear_svm"] = lambda: make_pipeline(
        SVC(
            kernel="linear",
            probability=True,
            class_weight="balanced",
            random_state=random_state,
        ),
        scale=True,
    )

    models["random_forest"] = lambda: make_pipeline(
        RandomForestClassifier(
            n_estimators=500,
            class_weight="balanced",
            random_state=random_state,
            n_jobs=-1,
        ),
        scale=False,
    )

    models["extra_trees"] = lambda: make_pipeline(
        ExtraTreesClassifier(
            n_estimators=700,
            class_weight="balanced",
            random_state=random_state,
            n_jobs=-1,
        ),
        scale=False,
    )

    models["hist_gb"] = lambda: make_pipeline(
        HistGradientBoostingClassifier(
            learning_rate=0.05,
            max_iter=300,
            l2_regularization=0.01,
            random_state=random_state,
        ),
        scale=False,
    )

    models["mlp"] = lambda: make_pipeline(
        MLPClassifier(
            hidden_layer_sizes=(256, 64),
            activation="relu",
            alpha=1e-4,
            learning_rate_init=1e-3,
            max_iter=800,
            early_stopping=True,
            random_state=random_state,
        ),
        scale=True,
    )

    # --------------------------------------------------------
    # Modelos opcionais
    # --------------------------------------------------------

    if USE_OPTIONAL_SOTA:

        if INSTALL_OPTIONAL_SOTA:
            for pkg in ["xgboost", "lightgbm", "catboost"]:
                try:
                    __import__(pkg)
                except Exception:
                    print(f"[INFO] Instalando pacote opcional: {pkg}")
                    subprocess.check_call([
                        sys.executable,
                        "-m",
                        "pip",
                        "install",
                        "-q",
                        pkg,
                    ])

        try:
            from xgboost import XGBClassifier

            models["xgboost"] = lambda: make_pipeline(
                XGBClassifier(
                    n_estimators=500,
                    max_depth=3,
                    learning_rate=0.03,
                    subsample=0.90,
                    colsample_bytree=0.90,
                    objective="multi:softprob",
                    eval_metric="mlogloss",
                    random_state=random_state,
                    n_jobs=-1,
                ),
                scale=False,
            )

        except Exception:
            print("[INFO] XGBoost não disponível.")

        try:
            from lightgbm import LGBMClassifier

            models["lightgbm"] = lambda: make_pipeline(
                LGBMClassifier(
                    n_estimators=700,
                    learning_rate=0.03,
                    num_leaves=31,
                    subsample=0.90,
                    colsample_bytree=0.90,
                    class_weight="balanced",
                    random_state=random_state,
                    n_jobs=-1,
                    verbose=-1,
                ),
                scale=False,
            )

        except Exception:
            print("[INFO] LightGBM não disponível.")

        try:
            from catboost import CatBoostClassifier

            models["catboost"] = lambda: make_pipeline(
                CatBoostClassifier(
                    iterations=700,
                    depth=4,
                    learning_rate=0.03,
                    loss_function="MultiClass",
                    random_seed=random_state,
                    verbose=False,
                    auto_class_weights="Balanced",
                ),
                scale=False,
            )

        except Exception:
            print("[INFO] CatBoost não disponível.")

    return models


# ============================================================
# TREINO COM RÓTULOS LOCAIS
# ============================================================

def class_names_to_global_ids(y_names, global_class_to_id):
    return np.array(
        [global_class_to_id[x] for x in y_names],
        dtype=int,
    )


def local_encode(y_global):
    """
    Converte ids globais para ids locais 0..K-1.

    Necessário para modelos que esperam labels consecutivos.
    """

    local_classes = sorted(np.unique(y_global).tolist())

    global_to_local = {
        g: i for i, g in enumerate(local_classes)
    }

    y_local = np.array(
        [global_to_local[g] for g in y_global],
        dtype=int,
    )

    return y_local, local_classes


def predict_proba_global_from_local_model(
    model,
    X,
    local_classes,
    n_global_classes,
):
    """
    O modelo foi treinado com labels locais.

    Aqui convertemos as probabilidades para o espaço global.
    """

    if hasattr(model, "predict_proba"):
        p_local = model.predict_proba(X)
    else:
        y_pred_local = model.predict(X)

        p_local = np.zeros(
            (len(y_pred_local), len(local_classes)),
            dtype=float,
        )

        for i, yy in enumerate(y_pred_local):
            p_local[i, int(yy)] = 1.0

    out = np.zeros(
        (p_local.shape[0], n_global_classes),
        dtype=float,
    )

    model_classes = model.named_steps["clf"].classes_

    for j, local_label in enumerate(model_classes):
        local_label = int(local_label)
        global_id = local_classes[local_label]
        out[:, global_id] = p_local[:, j]

    row_sum = out.sum(axis=1, keepdims=True)
    row_sum[row_sum <= 1e-12] = 1.0

    out = out / row_sum

    return out


def fit_model_local_labels(
    train_df,
    x_cols,
    model_factory,
    global_class_to_id,
):
    X_train = (
        train_df[x_cols]
        .apply(pd.to_numeric, errors="coerce")
        .values
        .astype(float)
    )

    y_global = class_names_to_global_ids(
        train_df["class_name"].values,
        global_class_to_id,
    )

    y_local, local_classes = local_encode(y_global)

    model = model_factory()
    model.fit(X_train, y_local)

    return model, local_classes


def predict_model_global(
    model,
    local_classes,
    test_df,
    x_cols,
    n_global_classes,
):
    X_test = (
        test_df[x_cols]
        .apply(pd.to_numeric, errors="coerce")
        .values
        .astype(float)
    )

    return predict_proba_global_from_local_model(
        model,
        X_test,
        local_classes,
        n_global_classes,
    )


# ============================================================
# MÉTRICAS
# ============================================================

def metrics_multiclass(y_true_names, y_pred_names, global_labels):
    """
    Métricas principais calculadas apenas nas classes presentes no teste.

    Isso evita penalizar AUDIO por não possuir AIRPLANE/BIRD ou IR+V por
    não possuir BACKGROUND.

    Também salva macro_f1_global apenas para auditoria.
    """

    y_true_names = np.asarray(y_true_names)
    y_pred_names = np.asarray(y_pred_names)

    present_labels = [
        c for c in global_labels
        if c in set(y_true_names)
    ]

    return {
        "accuracy": float(
            accuracy_score(y_true_names, y_pred_names)
        ),

        "balanced_accuracy": float(
            balanced_accuracy_score(y_true_names, y_pred_names)
        ),

        "macro_precision": float(
            precision_score(
                y_true_names,
                y_pred_names,
                labels=present_labels,
                average="macro",
                zero_division=0,
            )
        ),

        "macro_recall": float(
            recall_score(
                y_true_names,
                y_pred_names,
                labels=present_labels,
                average="macro",
                zero_division=0,
            )
        ),

        "macro_f1": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=present_labels,
                average="macro",
                zero_division=0,
            )
        ),

        "weighted_f1": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=present_labels,
                average="weighted",
                zero_division=0,
            )
        ),

        "macro_f1_global": float(
            f1_score(
                y_true_names,
                y_pred_names,
                labels=global_labels,
                average="macro",
                zero_division=0,
            )
        ),

        "labels_evaluated": "|".join(present_labels),
    }


def safe_stratified_split(df, random_state):
    counts = df["class_name"].value_counts()

    valid = counts[counts >= 2].index.tolist()

    df = df[df["class_name"].isin(valid)].copy()

    if df["class_name"].nunique() < 2:
        raise ValueError("Menos de duas classes com amostras suficientes.")

    train_df, test_df = train_test_split(
        df,
        test_size=TEST_SIZE,
        random_state=random_state,
        stratify=df["class_name"],
        shuffle=True,
    )

    return train_df.reset_index(drop=True), test_df.reset_index(drop=True)


# ============================================================
# AVALIAÇÃO DIRETA
# ============================================================

def evaluate_direct(
    train_df,
    test_df,
    x_cols,
    model_name,
    model_factory,
    combo_name,
    mode,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    model, local_classes = fit_model_local_labels(
        train_df,
        x_cols,
        model_factory,
        global_class_to_id,
    )

    p_global = predict_model_global(
        model,
        local_classes,
        test_df,
        x_cols,
        n_global,
    )

    pred_ids = np.argmax(p_global, axis=1)

    y_pred = np.array([
        global_classes[i] for i in pred_ids
    ])

    y_true = test_df["class_name"].values

    m = metrics_multiclass(
        y_true,
        y_pred,
        global_classes,
    )

    m.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": mode,
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(x_cols)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    preds = test_df[["sample_key", "class_name"]].copy()
    preds["repeat"] = repeat
    preds["combo"] = combo_name
    preds["mode"] = mode
    preds["model"] = model_name
    preds["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        preds[f"prob_{cls}"] = p_global[:, i]

    return m, preds


# ============================================================
# LATE FUSION
# ============================================================

def oof_sensor_predictions(
    train_df,
    test_df,
    sensor,
    model_factory,
    global_classes,
    global_class_to_id,
    random_state,
):
    fcols = get_feature_cols(train_df, sensor)
    has_col = f"has_{sensor}"

    n_global = len(global_classes)

    oof = np.zeros(
        (len(train_df), n_global),
        dtype=float,
    )

    test_prob = np.zeros(
        (len(test_df), n_global),
        dtype=float,
    )

    train_idx = np.where(train_df[has_col].values == 1)[0]
    test_idx = np.where(test_df[has_col].values == 1)[0]

    if len(train_idx) < 4:
        return oof, test_prob, 0.0

    tr_sensor = train_df.iloc[train_idx].copy()

    if tr_sensor["class_name"].nunique() < 2:
        return oof, test_prob, 0.0

    y_names = tr_sensor["class_name"].values

    y_global = class_names_to_global_ids(
        y_names,
        global_class_to_id,
    )

    min_count = pd.Series(y_names).value_counts().min()

    n_splits = min(
        MAX_CV_FOLDS,
        int(min_count),
    )

    if n_splits < 2:
        return oof, test_prob, 0.0

    X_sensor = (
        tr_sensor[fcols]
        .apply(pd.to_numeric, errors="coerce")
        .values
        .astype(float)
    )

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state,
    )

    for tr, va in skf.split(X_sensor, y_global):

        fold_df_train = tr_sensor.iloc[tr].copy()
        fold_df_val = tr_sensor.iloc[va].copy()

        model, local_classes = fit_model_local_labels(
            fold_df_train,
            fcols,
            model_factory,
            global_class_to_id,
        )

        p_val = predict_model_global(
            model,
            local_classes,
            fold_df_val,
            fcols,
            n_global,
        )

        original_val_idx = train_idx[va]

        oof[original_val_idx, :] = p_val

    model, local_classes = fit_model_local_labels(
        tr_sensor,
        fcols,
        model_factory,
        global_class_to_id,
    )

    if len(test_idx) > 0:
        te_sensor = test_df.iloc[test_idx].copy()

        p_test = predict_model_global(
            model,
            local_classes,
            te_sensor,
            fcols,
            n_global,
        )

        test_prob[test_idx, :] = p_test

    available_oof = oof[train_idx]

    pred_ids = np.argmax(available_oof, axis=1)

    y_pred = np.array([
        global_classes[i] for i in pred_ids
    ])

    y_true = tr_sensor["class_name"].values

    try:
        weight = balanced_accuracy_score(y_true, y_pred)
    except Exception:
        weight = 0.0

    return oof, test_prob, float(weight)


def normalize_rows(p):
    p = np.asarray(p, dtype=float)

    s = p.sum(axis=1, keepdims=True)
    s[s <= 1e-12] = 1.0

    return p / s


def evaluate_late_fusion(
    train_df,
    test_df,
    combo,
    combo_name,
    model_name,
    model_factory,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    sensor_test = {}
    sensor_weights = {}

    for sensor in combo:

        _, ptest, weight = oof_sensor_predictions(
            train_df,
            test_df,
            sensor,
            model_factory,
            global_classes,
            global_class_to_id,
            random_state=RANDOM_STATE + repeat,
        )

        sensor_test[sensor] = ptest
        sensor_weights[sensor] = weight

    y_true = test_df["class_name"].values

    # --------------------------------------------------------
    # Late fusion por média simples
    # --------------------------------------------------------

    p_sum = np.zeros(
        (len(test_df), n_global),
        dtype=float,
    )

    count = np.zeros(
        (len(test_df), 1),
        dtype=float,
    )

    for sensor in combo:
        has = (
            test_df[f"has_{sensor}"]
            .values
            .reshape(-1, 1)
            .astype(float)
        )

        p_sum += sensor_test[sensor] * has
        count += has

    count[count <= 1e-12] = 1.0

    p_mean = normalize_rows(p_sum / count)

    y_pred = np.array([
        global_classes[i] for i in np.argmax(p_mean, axis=1)
    ])

    m_mean = metrics_multiclass(
        y_true,
        y_pred,
        global_classes,
    )

    m_mean.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_mean",
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(combo)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    pred_mean = test_df[["sample_key", "class_name"]].copy()
    pred_mean["repeat"] = repeat
    pred_mean["combo"] = combo_name
    pred_mean["mode"] = "late_mean"
    pred_mean["model"] = model_name
    pred_mean["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        pred_mean[f"prob_{cls}"] = p_mean[:, i]

    # --------------------------------------------------------
    # Late fusion ponderada por OOF
    # --------------------------------------------------------

    p_wsum = np.zeros(
        (len(test_df), n_global),
        dtype=float,
    )

    w_sum = np.zeros(
        (len(test_df), 1),
        dtype=float,
    )

    for sensor in combo:
        has = (
            test_df[f"has_{sensor}"]
            .values
            .reshape(-1, 1)
            .astype(float)
        )

        w = max(sensor_weights.get(sensor, 0.0), 1e-6)

        p_wsum += sensor_test[sensor] * has * w
        w_sum += has * w

    w_sum[w_sum <= 1e-12] = 1.0

    p_weighted = normalize_rows(p_wsum / w_sum)

    y_pred_w = np.array([
        global_classes[i] for i in np.argmax(p_weighted, axis=1)
    ])

    m_weighted = metrics_multiclass(
        y_true,
        y_pred_w,
        global_classes,
    )

    m_weighted.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_weighted_oof",
        "model": model_name,
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(len(combo)),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    pred_weighted = test_df[["sample_key", "class_name"]].copy()
    pred_weighted["repeat"] = repeat
    pred_weighted["combo"] = combo_name
    pred_weighted["mode"] = "late_weighted_oof"
    pred_weighted["model"] = model_name
    pred_weighted["pred_class"] = y_pred_w

    for i, cls in enumerate(global_classes):
        pred_weighted[f"prob_{cls}"] = p_weighted[:, i]

    return [m_mean, m_weighted], [pred_mean, pred_weighted]


def evaluate_late_stacking(
    train_df,
    test_df,
    combo,
    combo_name,
    base_models,
    repeat,
    global_classes,
    global_class_to_id,
):
    n_global = len(global_classes)

    train_parts = []
    test_parts = []

    for model_name, factory in base_models.items():
        for sensor in combo:

            oof, ptest, _ = oof_sensor_predictions(
                train_df,
                test_df,
                sensor,
                factory,
                global_classes,
                global_class_to_id,
                random_state=RANDOM_STATE + repeat,
            )

            train_parts.append(oof)
            test_parts.append(ptest)

    for sensor in combo:
        train_parts.append(
            train_df[[f"has_{sensor}"]].values.astype(float)
        )

        test_parts.append(
            test_df[[f"has_{sensor}"]].values.astype(float)
        )

    X_train_meta = np.hstack(train_parts)
    X_test_meta = np.hstack(test_parts)

    y_global = class_names_to_global_ids(
        train_df["class_name"].values,
        global_class_to_id,
    )

    y_local, local_classes = local_encode(y_global)

    meta = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            random_state=RANDOM_STATE + repeat,
        )),
    ])

    meta.fit(X_train_meta, y_local)

    p_global = predict_proba_global_from_local_model(
        meta,
        X_test_meta,
        local_classes,
        n_global,
    )

    y_pred = np.array([
        global_classes[i] for i in np.argmax(p_global, axis=1)
    ])

    y_true = test_df["class_name"].values

    m = metrics_multiclass(
        y_true,
        y_pred,
        global_classes,
    )

    m.update({
        "repeat": repeat,
        "combo": combo_name,
        "mode": "late_stacking",
        "model": "logreg_meta",
        "n_samples": int(len(train_df) + len(test_df)),
        "n_train": int(len(train_df)),
        "n_test": int(len(test_df)),
        "n_features": int(X_train_meta.shape[1]),
        "classes_in_test": "|".join(sorted(np.unique(y_true))),
    })

    preds = test_df[["sample_key", "class_name"]].copy()
    preds["repeat"] = repeat
    preds["combo"] = combo_name
    preds["mode"] = "late_stacking"
    preds["model"] = "logreg_meta"
    preds["pred_class"] = y_pred

    for i, cls in enumerate(global_classes):
        preds[f"prob_{cls}"] = p_global[:, i]

    return m, preds


# ============================================================
# SUMÁRIO / PLOTS
# ============================================================

def summarize_metrics(metrics_df):
    metric_cols = [
        "accuracy",
        "balanced_accuracy",
        "macro_precision",
        "macro_recall",
        "macro_f1",
        "macro_f1_global",
        "weighted_f1",
    ]

    agg = {}

    for m in metric_cols:
        if m in metrics_df.columns:
            agg[f"{m}_mean"] = (m, "mean")
            agg[f"{m}_std"] = (m, "std")

    agg["n_samples_mean"] = ("n_samples", "mean")
    agg["n_test_mean"] = ("n_test", "mean")
    agg["n_features_mean"] = ("n_features", "mean")

    out = (
        metrics_df
        .groupby(["combo", "mode", "model"])
        .agg(**agg)
        .reset_index()
        .sort_values(
            [
                "balanced_accuracy_mean",
                "macro_f1_mean",
                "weighted_f1_mean",
            ],
            ascending=False,
        )
    )

    return out


def save_confusion_matrix(y_true, y_pred, labels, out_path, title):
    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=labels,
    )

    plt.figure(figsize=(8, 7))
    plt.imshow(cm)
    plt.title(title)
    plt.xticks(
        range(len(labels)),
        labels,
        rotation=45,
        ha="right",
    )
    plt.yticks(
        range(len(labels)),
        labels,
    )
    plt.xlabel("Predito")
    plt.ylabel("Real")

    for i in range(len(labels)):
        for j in range(len(labels)):
            plt.text(
                j,
                i,
                str(cm[i, j]),
                ha="center",
                va="center",
            )

    plt.tight_layout()
    plt.savefig(out_path, dpi=200)
    plt.close()


# ============================================================
# MAIN
# ============================================================

def main():

    print("=" * 80)
    print("CLASSIFICAÇÃO MULTICLASSE COM IR + V + AUDIO")
    print("VERSÃO CORRIGIDA: MÉTRICAS POR CLASSES PRESENTES")
    print("=" * 80)

    tables = load_sensor_tables()

    class_availability = print_class_availability(tables)

    global_classes = sorted(
        class_availability["class_name"].unique().tolist()
    )

    global_class_to_id = {
        c: i for i, c in enumerate(global_classes)
    }

    print("\nClasses globais:")

    for i, cls in enumerate(global_classes):
        print(f"  {i}: {cls}")

    with open(
        OUT_DIR / "global_classes.json",
        "w",
        encoding="utf-8",
    ) as f:
        json.dump(global_classes, f, indent=2)

    all_metrics = []
    all_preds = []
    combo_sizes = []

    for combo in COMBOS:

        combo_name = "+".join(combo)

        print("\n" + "#" * 80)
        print(f"COMBINAÇÃO: {combo_name}")
        print("#" * 80)

        df_combo = build_union_dataset(tables, combo)

        counts = df_combo["class_name"].value_counts().sort_index()

        print("Amostras:", len(df_combo))
        print("Classes:")
        print(counts.to_string())

        combo_sizes.append({
            "combo": combo_name,
            "n_samples": int(len(df_combo)),
            "classes": "|".join(counts.index.tolist()),
            **{
                f"n_{cls}": int(counts.get(cls, 0))
                for cls in global_classes
            },
        })

        if df_combo["class_name"].nunique() < 2:
            print("[SKIP] Apenas uma classe.")
            continue

        for repeat in range(N_REPEATS):

            seed = RANDOM_STATE + repeat

            print(f"\n{combo_name} | repetição {repeat + 1}/{N_REPEATS}")

            try:
                train_df, test_df = safe_stratified_split(
                    df_combo,
                    random_state=seed,
                )

            except Exception as e:
                print("[SKIP]", e)
                continue

            models = model_zoo(random_state=seed)

            # ------------------------------------------------
            # 1) Unimodal
            # ------------------------------------------------

            for sensor in combo:

                x_cols = get_feature_cols(train_df, sensor)

                tr_s = train_df[
                    train_df[f"has_{sensor}"] == 1
                ].copy()

                te_s = test_df[
                    test_df[f"has_{sensor}"] == 1
                ].copy()

                if len(tr_s) < 5 or len(te_s) < 2:
                    continue

                if tr_s["class_name"].nunique() < 2:
                    continue

                for model_name, factory in models.items():

                    try:
                        m, p = evaluate_direct(
                            tr_s,
                            te_s,
                            x_cols,
                            model_name,
                            factory,
                            combo_name,
                            f"unimodal_{sensor}",
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                    except Exception as e:
                        print(
                            f"[ERRO] {combo_name} "
                            f"unimodal {sensor} {model_name}: {e}"
                        )

            # ------------------------------------------------
            # 2) Early fusion com sensores ausentes
            # ------------------------------------------------

            if len(combo) >= 2:

                x_cols = get_combo_feature_cols(train_df, combo)

                for model_name, factory in models.items():

                    try:
                        m, p = evaluate_direct(
                            train_df,
                            test_df,
                            x_cols,
                            model_name,
                            factory,
                            combo_name,
                            "early_concat_missing",
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                    except Exception as e:
                        print(
                            f"[ERRO] {combo_name} "
                            f"early {model_name}: {e}"
                        )

            # ------------------------------------------------
            # 3) Late fusion por modelo
            # ------------------------------------------------

            if len(combo) >= 2:

                for model_name, factory in models.items():

                    try:
                        metrics, preds = evaluate_late_fusion(
                            train_df,
                            test_df,
                            combo,
                            combo_name,
                            model_name,
                            factory,
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.extend(metrics)
                        all_preds.extend(preds)

                    except Exception as e:
                        print(
                            f"[ERRO] {combo_name} "
                            f"late {model_name}: {e}"
                        )

                # ------------------------------------------------
                # 4) Late stacking
                # ------------------------------------------------

                try:
                    stacking_models = {
                        k: v for k, v in models.items()
                        if k in [
                            "logreg_l2",
                            "linear_svm",
                            "random_forest",
                            "extra_trees",
                            "hist_gb",
                            "lightgbm",
                            "catboost",
                        ]
                    }

                    if len(stacking_models) > 0:

                        m, p = evaluate_late_stacking(
                            train_df,
                            test_df,
                            combo,
                            combo_name,
                            stacking_models,
                            repeat,
                            global_classes,
                            global_class_to_id,
                        )

                        all_metrics.append(m)
                        all_preds.append(p)

                except Exception as e:
                    print(f"[ERRO] {combo_name} stacking: {e}")

    # ========================================================
    # SALVAR
    # ========================================================

    combo_sizes_df = pd.DataFrame(combo_sizes)

    combo_sizes_path = OUT_DIR / "combination_sizes_multiclass.csv"

    combo_sizes_df.to_csv(
        combo_sizes_path,
        index=False,
        encoding="utf-8-sig",
    )

    metrics_df = pd.DataFrame(all_metrics)

    metrics_path = OUT_DIR / "metrics_all_repeats_multiclass.csv"

    metrics_df.to_csv(
        metrics_path,
        index=False,
        encoding="utf-8-sig",
    )

    summary_df = (
        summarize_metrics(metrics_df)
        if not metrics_df.empty
        else pd.DataFrame()
    )

    summary_path = OUT_DIR / "metrics_summary_multiclass.csv"

    summary_df.to_csv(
        summary_path,
        index=False,
        encoding="utf-8-sig",
    )

    preds_df = (
        pd.concat(all_preds, ignore_index=True)
        if all_preds
        else pd.DataFrame()
    )

    preds_path = OUT_DIR / "predictions_all_multiclass.csv"

    preds_df.to_csv(
        preds_path,
        index=False,
        encoding="utf-8-sig",
    )

    if not summary_df.empty and not preds_df.empty:

        best = summary_df.iloc[0]

        best_preds = preds_df[
            (preds_df["combo"] == best["combo"]) &
            (preds_df["mode"] == best["mode"]) &
            (preds_df["model"] == best["model"])
        ].copy()

        cm_path = OUT_DIR / "confusion_matrix_best_model.png"

        save_confusion_matrix(
            best_preds["class_name"].values,
            best_preds["pred_class"].values,
            global_classes,
            cm_path,
            f"{best['combo']} | {best['mode']} | {best['model']}",
        )

    # ========================================================
    # PRINT FINAL
    # ========================================================

    print("\n" + "=" * 80)
    print("TAMANHO DAS COMBINAÇÕES")
    print("=" * 80)

    if not combo_sizes_df.empty:
        print(combo_sizes_df.to_string(index=False))
    else:
        print("[AVISO] Nenhuma combinação avaliada.")

    print("\n" + "=" * 80)
    print("TOP RESULTADOS MULTICLASSE")
    print("=" * 80)

    if not summary_df.empty:

        show_cols = [
            "combo",
            "mode",
            "model",
            "balanced_accuracy_mean",
            "balanced_accuracy_std",
            "macro_f1_mean",
            "macro_f1_std",
            "macro_f1_global_mean",
            "macro_f1_global_std",
            "weighted_f1_mean",
            "weighted_f1_std",
            "n_samples_mean",
            "n_features_mean",
        ]

        show_cols = [
            c for c in show_cols
            if c in summary_df.columns
        ]

        print(
            summary_df[show_cols]
            .head(40)
            .to_string(index=False)
        )

    else:
        print("[AVISO] Nenhum resultado gerado.")

    print("\n" + "=" * 80)
    print("ARQUIVOS SALVOS")
    print("=" * 80)
    print(f"[OK] {combo_sizes_path}")
    print(f"[OK] {metrics_path}")
    print(f"[OK] {summary_path}")
    print(f"[OK] {preds_path}")
    print(f"[OK] {OUT_DIR / 'global_classes.json'}")
    print(f"[OK] Diretório:\n{OUT_DIR}")


if __name__ == "__main__":
    main()
```
